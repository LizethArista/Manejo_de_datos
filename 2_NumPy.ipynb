# Introducción a NumPy

NumPy es una biblioteca fundamental para manejo de datos en Python

Nos permite trabajar con arreglos numéricos de forma más eficiente que las listas normales


- Arreglos de una y varias dimensiones
- Formas y tipos de datos
- Indexación y rebanado
- Operaciones vectorizadas
- Filtros booleanos
- Broadcast
- Estadísticas básicas

````python
import numpy as np

print("Versión de NumPy:", np.__version__)
````

## ¿Por qué usar NumPy?

Las listas de Python son útiles, pero para operaciones numéricas grandes no son tan eficientes


Una lista es una estructura de datos flexible de Python, mientras que un arreglo de NumPy está pensado para hacer operaciones numéricas eficientes


NumPy ofrece:

- Arreglos homogéneos
- Operaciones vectorizadas
- Menos ciclos `for`
- Herramientas para álgebra y estadística


| Característica | Lista de Python | Arreglo NumPy |
|---|---|---|
| Tipo de datos | Puede mezclar tipos: enteros, textos, booleanos, etc. | Normalmente todos los elementos son del mismo tipo |
| Eficiencia | Menos eficiente para cálculos grandes | Más eficiente para operaciones numéricas |
| Operaciones | Se usan ciclos o comprensiones | Permite operaciones vectorizadas |
| Dimensiones | Normalmente se maneja como 1D, aunque puede anidarse | Puede ser 1D, 2D, 3D, etc. |
| Funciones matemáticas | No tiene funciones matemáticas integradas | Tiene muchas: suma, promedio, máximo, etc. |
| Uso principal | Estructura de datos general | Cálculo numérico y manejo de datos |

````python
lista = [1, 2, 3, 4]
arreglo = np.array(lista)

print("Lista:", lista)
print("Arreglo:", arreglo)

print("Tipo de lista:", type(lista))
print("Tipo de arreglo:", type(arreglo))

print("dtype:", arreglo.dtype)
print("shape:", arreglo.shape)
````

````python
print(lista * 2)
print(arreglo * 2)
````

## Creación de arreglos

NumPy tiene varias funciones para crear arreglos:

| Función | Descripción |
|---|---|
| `np.array()` | Crea un arreglo desde una lista |
| `np.zeros()` | Crea arreglo lleno de ceros |
| `np.ones()` | Crea arreglo lleno de unos |
| `np.arange()` | Crea secuencias con paso |
| `np.linspace()` | Crea valores igualmente espaciados |
| `np.random.default_rng()` | Generador aleatorio |

````python
rng = np.random.default_rng(49)

a = np.array([10, 20, 30])
print("a:", a)

````

````python
b = np.zeros((2, 3), dtype=int)  # crea una matriz de 2 filas y 3 columnas
print("b:\n", b)
````

````python
c = np.ones(5)
print("c:", c)
````

````python
d = np.arange(1, 11, 2)  #genera números de 1 a 10 con paso 2
print("d:", d)
````

````python
e = np.linspace(0, 1, 5)
print("e:", e)
````

````python
f = rng.integers(1, 101, size=6) # genera 6 enteros aleatorios entre 1 y 100
print("f:", f)
````

## Atributos de un arreglo

Los arreglos tienen propiedades importantes:

- `shape`: forma o dimensiones
- `ndim`: número de dimensiones
- `size`: cantidad total de elementos
- `dtype`: tipo de dato

````python
matriz = np.arange(1, 13).reshape(3, 4)

print(matriz)

print("ndim:", matriz.ndim)
print("shape:", matriz.shape)
print("size:", matriz.size)
print("dtype:", matriz.dtype)
````

## Cambiar forma con `reshape`

Podemos reorganizar los elementos de un arreglo sin cambiar sus datos.

````python
numeros = np.arange(1, 13)

print("Original:", numeros)

matriz_3x4 = numeros.reshape(3, 4)
print("Matriz 3x4:\n", matriz_3x4)

matriz_4x3 = numeros.reshape(4, 3)
print("Matriz 4x3:\n", matriz_4x3)

print("Aplanado:", matriz_3x4.flatten())
````

## Indexación y rebanado

Podemos acceder a elementos usando índices.

En arreglos de una dimensión:

```python
arreglo[posicion]
arreglo[inicio:fin:paso]


```
en una matriz
```python
matriz[fila, columna]

````python
numeros = np.arange(10, 110, 10)

print(numeros)

print("Primero:", numeros[0])
print("Último:", numeros[-1])
print("Rebanada 2:6:", numeros[2:6])
print("Cada 3 posiciones:", numeros[::3])
````

````python
notas = np.array([
    [8, 7, 9],
    [10, 6, 8],
    [5, 9, 10]
])

print(notas)
print("Fila 0, columna 0:", notas[0, 0])

````

````python
print(notas)
print("Fila completa 1:", notas[1])
````

````python
print(notas)
print("Columna 2 completa:", notas[:, 2])
````

````python
print(notas)
print("Submatriz filas 0-1, columnas 1-2:\n", notas[0:2, 1:3])
````

## Filtros booleanos

Los filtros booleanos permiten seleccionar elementos que cumplen una condición.

Ejemplo:

```python
arreglo[arreglo > 5]

````python
calificaciones = np.array([5, 7, 8, 10, 4, 9])

aprobo = calificaciones >= 7

print("Calificaciones:", calificaciones)
print("¿Aprobó?:", aprobo)
print("Aprobadas:", calificaciones[aprobo])
print("Pares:", calificaciones[calificaciones % 2 == 0])
````



## Operaciones vectorizadas

Con NumPy podemos hacer operaciones elemento por elemento sin escribir ciclos

````python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print("a + b:", a + b)
print("a - b:", a - b)
print("a * b:", a * b)
print("a / b:", a / b)
print("a ** 2:", a ** 2)
print("sqrt(a):", np.sqrt(a))
````

## Broadcasting

Broadcasting permite combinar arreglos de diferentes formas cuando las dimensiones son compatibles

Por ejemplo, sumar un número a toda una matriz

````python
m = np.arange(6).reshape(2, 3)

print("Matriz:\n", m)

print("Sumar 100:\n", m + 100)  # m + 100: suma 100 a todos los elementos

print("Sumar fila:\n", m + np.array([10, 20, 30])) #suma una fila compatible

print("Sumar columna:\n", m + np.array([[100], [200]])) #suma valores por columna
````

## Estadísticas y agregaciones

NumPy incluye funciones para calcular:

- Suma.
- Promedio.
- Máximo.
- Mínimo.
- Índice del máximo.
- Índice del mínimo.

````python
ventas = rng.integers(100, 500, size=(4, 5))

print("Ventas:\n", ventas)

print("Suma total:", ventas.sum())
print("Promedio:", ventas.mean())
print("Máximo:", ventas.max())
print("Mínimo:", ventas.min())
print("Índice del máximo:", ventas.argmax())
````

## Uso de `axis`

Cuando trabajamos con matrices:

- `axis=0`: opera por columnas.
- `axis=1`: opera por filas.

Si `ventas` tiene forma `(4, 5)`:

- 4 filas pueden ser tiendas.
- 5 columnas pueden ser días.

````python
print("Ventas:\n", ventas)

print("Total por día:", ventas.sum(axis=0))
print("Total por tienda:", ventas.sum(axis=1))

print("Promedio por día:", ventas.mean(axis=0))
print("Promedio por tienda:", ventas.mean(axis=1))
````

## Unir arreglos

Podemos combinar arreglos usando:

- `np.hstack()`: unir horizontalmente
- `np.vstack()`: unir verticalmente

````python
turno1 = np.array([8, 9, 7])
turno2 = np.array([10, 6, 9])

print("Horizontal:", np.hstack([turno1, turno2]))
print("Vertical:\n", np.vstack([turno1, turno2]))
````

Supongamos que tenemos calificaciones de tres estudiantes en tres materias

Queremos:

- Calcular promedio por estudiante
- Saber quién tiene el mejor promedio
- Contar cuántas materias aprobó cada estudiante

````python
alumnos = ["Ana", "Beto", "Carla"]
materias = ["Matemáticas", "Física", "Química"]

calificaciones = np.array([
    [8, 7, 9],
    [6, 8, 7],
    [10, 10, 9]
])

print("Calificaciones:\n", calificaciones)

promedios = calificaciones.mean(axis=1)

print("Promedios:", promedios)

````

````python
mejor_indice = np.argmax(promedios)

print("Mejor promedio:", alumnos[mejor_indice])
print("Promedio:", promedios[mejor_indice])
````

````python
aprobadas = calificaciones >= 7

print("Materias aprobadas por estudiante:\n", aprobadas)
print("Cantidad de materias aprobadas:", aprobadas.sum(axis=1))
````
