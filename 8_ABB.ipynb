## ¿Qué es un Árbol Binario de Búsqueda (ABB)?

Un **Árbol Binario de Búsqueda** es una estructura de datos jerárquica donde cada nodo tiene **como máximo dos hijos** (izquierdo y derecho) y cumple la siguiente **propiedad **:

> Para cualquier nodo `N`:
> - Todos los valores en su **subárbol izquierdo** son **menores** que `N.valor`.
> - Todos los valores en su **subárbol derecho** son **mayores** que `N.valor`.


```python
    10
   /  \
  5    15
 / \   / \
2   7 12  20
```


**Cumple ABB:** todo lo de la izquierda de 10 es menor, todo lo de la derecha es mayor

**No cumple ABB:**

```python
    10
   /  \
  15    5    ← invertido
```

## Complejidades de un ABB

Sea `n` la cantidad de nodos y `h` la altura del árbol:

| Operación       | Caso promedio | Peor caso |
|-----------------|:-------------:|:---------:|
| Búsqueda        | O(log n)      | O(n)      |
| Inserción       | O(log n)      | O(n)      |
| Eliminación     | O(log n)      | O(n)      |
| Recorrido       | O(n)          | O(n)      |

**El peor caso** ocurre cuando el árbol está **completamente desbalanceado** (se comporta como una lista enlazada)


```python
# ============================================================
# DEFINICIÓN DEL NODO Y DEL ÁRBOL BINARIO DE BÚSQUEDA
# ============================================================

class Nodo:
    """Representa un nodo del árbol binario."""
    def __init__(self, valor):
        self.valor = valor     # dato almacenado
        self.izq = None        # hijo izquierdo
        self.der = None        # hijo derecho


class ABB:
    """Árbol Binario de Búsqueda."""
    def __init__(self):
        self.raiz = None

    # ---------- INSERCIÓN ----------
    def insertar(self, valor):
        """Inserta un valor manteniendo la propiedad del ABB."""
        if self.raiz is None:
            self.raiz = Nodo(valor)
        else:
            self._insertar_rec(self.raiz, valor)

    def _insertar_rec(self, nodo, valor):
        if valor < nodo.valor:
            if nodo.izq is None:
                nodo.izq = Nodo(valor)
            else:
                self._insertar_rec(nodo.izq, valor)
        elif valor > nodo.valor:
            if nodo.der is None:
                nodo.der = Nodo(valor)
            else:
                self._insertar_rec(nodo.der, valor)
        # Si valor == nodo.valor: no se inserta (sin duplicados)

    # ---------- BÚSQUEDA ----------
    def buscar(self, valor):
        """Retorna True si el valor existe en el árbol."""
        return self._buscar_rec(self.raiz, valor)

    def _buscar_rec(self, nodo, valor):
        if nodo is None:
            return False
        if valor == nodo.valor:
            return True
        if valor < nodo.valor:
            return self._buscar_rec(nodo.izq, valor)
        return self._buscar_rec(nodo.der, valor)


# --- Construyamos el árbol del ejemplo ---
#         10
#        /  \
#       5    15
#      / \   / \
#     2   7 12  20

arbol = ABB()
for v in [10, 5, 15, 2, 7, 12, 20]:
    arbol.insertar(v)

print("¿Existe el 7?", arbol.buscar(7))   # True
print("¿Existe el 8?", arbol.buscar(8))   # False
```

## Los tres recorridos

Los recorridos visitan **todos los nodos** del árbol. La diferencia está en **cuándo se procesa la raíz**.

###  Preorden (Raíz – Izquierda – Derecha)
Se procesa la raíz **antes** de visitar los hijos.

Útil para **copiar/clonar** el árbol

### Inorden (Izquierda – Raíz – Derecha)
Se procesa la raíz **entre** los hijos.

En un ABB, el inorden devuelve los valores **ordenados de menor a mayor**


###  Postorden (Izquierda – Derecha – Raíz)
Se procesa la raíz **después** de visitar los hijos.

Útil para **eliminar** el árbol (liberar memoria) o evaluar expresiones


```python
    10
   /  \
  5    15
 / \   / \
2   7 12  20
```

Preorden → 10, 5, 2, 7, 15, 12, 20
Inorden → 2, 5, 7, 10, 12, 15, 20 ← ordenado
Postorden → 2, 7, 5, 12, 20, 15, 10


**Complejidad:** O(n) en todos los casos, ya que visitamos cada nodo exactamente una vez

**Espacio auxiliar:** O(h) por la pila de recursión, donde h es la altura

```python

# RECORRIDOS: PREORDEN, INORDEN, POSTORDEN


def preorden(nodo, resultado=None):
    """
    Preorden: Raíz → Izquierda → Derecha
    Complejidad: O(n) tiempo, O(h) espacio (pila de recursión)
    """
    if resultado is None:
        resultado = []
    if nodo is not None:
        resultado.append(nodo.valor)       # 1) procesar raíz
        preorden(nodo.izq, resultado)      # 2) recorrer subárbol izq
        preorden(nodo.der, resultado)      # 3) recorrer subárbol der
    return resultado


def inorden(nodo, resultado=None):
    """
    Inorden: Izquierda → Raíz → Derecha
    En un ABB devuelve los valores ORDENADOS
    Complejidad: O(n) tiempo, O(h) espacio
    """
    if resultado is None:
        resultado = []
    if nodo is not None:
        inorden(nodo.izq, resultado)       # 1) recorrer subárbol izq
        resultado.append(nodo.valor)       # 2) procesar raíz
        inorden(nodo.der, resultado)       # 3) recorrer subárbol der
    return resultado


def postorden(nodo, resultado=None):
    """
    Postorden: Izquierda → Derecha → Raíz
    Útil para eliminar el árbol o evaluar expresiones
    Complejidad: O(n) tiempo, O(h) espacio
    """
    if resultado is None:
        resultado = []
    if nodo is not None:
        postorden(nodo.izq, resultado)     # 1) recorrer subárbol izq
        postorden(nodo.der, resultado)     # 2) recorrer subárbol der
        resultado.append(nodo.valor)       # 3) procesar raíz
    return resultado


# --- Probemos con nuestro árbol ---
print("Preorden :", preorden(arbol.raiz))
print("Inorden  :", inorden(arbol.raiz))
print("Postorden:", postorden(arbol.raiz))
```

```python
# INORDEN ITERATIVO (usando pila explícita)
# Evita el uso de recursión, es útil cuando h es muy grande


def inorden_iterativo(raiz):
    """Inorden sin recursión, usando una pila."""
    resultado = []
    pila = []
    actual = raiz

    while actual is not None or pila:
        # Llegar al nodo más a la izquierda posible
        while actual is not None:
            pila.append(actual)
            actual = actual.izq

        # Procesar el nodo
        actual = pila.pop()
        resultado.append(actual.valor)

        # Pasar al subárbol derecho
        actual = actual.der

    return resultado

print("Inorden iterativo:", inorden_iterativo(arbol.raiz))
```



| Concepto | Idea clave |
|----------|-----------|
| ABB | Izq < raíz < Der |
| Preorden | R-I-D → copiar árbol |
| Inorden | I-R-D → valores ordenados |
| Postorden | I-D-R → eliminar árbol |
| Complejidad recorridos | O(n) tiempo, O(h) espacio |
