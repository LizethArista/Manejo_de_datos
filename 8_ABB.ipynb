## ¿Qué es un Árbol Binario de Búsqueda (ABB)?

Un **Árbol Binario de Búsqueda** es una estructura de datos jerárquica donde cada nodo tiene **como máximo dos hijos** (izquierdo y derecho) y cumple la siguiente **propiedad **:

> Para cualquier nodo `N`:
> - Todos los valores en su **subárbol izquierdo** son **menores** que `N.valor`.
> - Todos los valores en su **subárbol derecho** son **mayores** que `N.valor`.


```python
    10
   /  \
  5    15
 / \   / \
2   7 12  20
```


**Cumple ABB:** todo lo de la izquierda de 10 es menor, todo lo de la derecha es mayor

**No cumple ABB:**

```python
    10
   /  \
  15    5    ← invertido
```

## Complejidades de un ABB

Sea `n` la cantidad de nodos y `h` la altura del árbol:

| Operación       | Caso promedio | Peor caso |
|-----------------|:-------------:|:---------:|
| Búsqueda        | O(log n)      | O(n)      |
| Inserción       | O(log n)      | O(n)      |
| Eliminación     | O(log n)      | O(n)      |
| Recorrido       | O(n)          | O(n)      |

**El peor caso** ocurre cuando el árbol está **completamente desbalanceado** (se comporta como una lista enlazada)


```python
# ============================================================
# DEFINICIÓN DEL NODO Y DEL ÁRBOL BINARIO DE BÚSQUEDA
# ============================================================

class Nodo:
    """Representa un nodo del árbol binario."""
    def __init__(self, valor):
        self.valor = valor     # dato almacenado
        self.izq = None        # hijo izquierdo
        self.der = None        # hijo derecho


class ABB:
    """Árbol Binario de Búsqueda."""
    def __init__(self):
        self.raiz = None

    # ---------- INSERCIÓN ----------
    def insertar(self, valor):
        """Inserta un valor manteniendo la propiedad del ABB."""
        if self.raiz is None:
            self.raiz = Nodo(valor)
        else:
            self._insertar_rec(self.raiz, valor)

    def _insertar_rec(self, nodo, valor):
        if valor < nodo.valor:
            if nodo.izq is None:
                nodo.izq = Nodo(valor)
            else:
                self._insertar_rec(nodo.izq, valor)
        elif valor > nodo.valor:
            if nodo.der is None:
                nodo.der = Nodo(valor)
            else:
                self._insertar_rec(nodo.der, valor)
        # Si valor == nodo.valor: no se inserta (sin duplicados)

    # ---------- BÚSQUEDA ----------
    def buscar(self, valor):
        """Retorna True si el valor existe en el árbol."""
        return self._buscar_rec(self.raiz, valor)

    def _buscar_rec(self, nodo, valor):
        if nodo is None:
            return False
        if valor == nodo.valor:
            return True
        if valor < nodo.valor:
            return self._buscar_rec(nodo.izq, valor)
        return self._buscar_rec(nodo.der, valor)


# --- Construyamos el árbol del ejemplo ---
#         10
#        /  \
#       5    15
#      / \   / \
#     2   7 12  20

arbol = ABB()
for v in [10, 5, 15, 2, 7, 12, 20]:
    arbol.insertar(v)

print("¿Existe el 7?", arbol.buscar(7))   # True
print("¿Existe el 8?", arbol.buscar(8))   # False
```

## Los tres recorridos

Los recorridos visitan **todos los nodos** del árbol. La diferencia está en **cuándo se procesa la raíz**.

###  Preorden (Raíz – Izquierda – Derecha)
Se procesa la raíz **antes** de visitar los hijos.

Útil para **copiar/clonar** el árbol

### Inorden (Izquierda – Raíz – Derecha)
Se procesa la raíz **entre** los hijos.

En un ABB, el inorden devuelve los valores **ordenados de menor a mayor**


###  Postorden (Izquierda – Derecha – Raíz)
Se procesa la raíz **después** de visitar los hijos.

Útil para **eliminar** el árbol (liberar memoria) o evaluar expresiones


```python
    10
   /  \
  5    15
 / \   / \
2   7 12  20
```

Preorden → 10, 5, 2, 7, 15, 12, 20
Inorden → 2, 5, 7, 10, 12, 15, 20 ← ordenado
Postorden → 2, 7, 5, 12, 20, 15, 10


**Complejidad:** O(n) en todos los casos, ya que visitamos cada nodo exactamente una vez

**Espacio auxiliar:** O(h) por la pila de recursión, donde h es la altura

```python

# RECORRIDOS: PREORDEN, INORDEN, POSTORDEN


def preorden(nodo, resultado=None):
    """
    Preorden: Raíz → Izquierda → Derecha
    Complejidad: O(n) tiempo, O(h) espacio (pila de recursión)
    """
    if resultado is None:
        resultado = []
    if nodo is not None:
        resultado.append(nodo.valor)       # 1) procesar raíz
        preorden(nodo.izq, resultado)      # 2) recorrer subárbol izq
        preorden(nodo.der, resultado)      # 3) recorrer subárbol der
    return resultado


def inorden(nodo, resultado=None):
    """
    Inorden: Izquierda → Raíz → Derecha
    En un ABB devuelve los valores ORDENADOS
    Complejidad: O(n) tiempo, O(h) espacio
    """
    if resultado is None:
        resultado = []
    if nodo is not None:
        inorden(nodo.izq, resultado)       # 1) recorrer subárbol izq
        resultado.append(nodo.valor)       # 2) procesar raíz
        inorden(nodo.der, resultado)       # 3) recorrer subárbol der
    return resultado


def postorden(nodo, resultado=None):
    """
    Postorden: Izquierda → Derecha → Raíz
    Útil para eliminar el árbol o evaluar expresiones
    Complejidad: O(n) tiempo, O(h) espacio
    """
    if resultado is None:
        resultado = []
    if nodo is not None:
        postorden(nodo.izq, resultado)     # 1) recorrer subárbol izq
        postorden(nodo.der, resultado)     # 2) recorrer subárbol der
        resultado.append(nodo.valor)       # 3) procesar raíz
    return resultado


# --- Probemos con nuestro árbol ---
print("Preorden :", preorden(arbol.raiz))
print("Inorden  :", inorden(arbol.raiz))
print("Postorden:", postorden(arbol.raiz))
```

```python
# INORDEN ITERATIVO (usando pila explícita)
# Evita el uso de recursión, es útil cuando h es muy grande


def inorden_iterativo(raiz):
    """Inorden sin recursión, usando una pila."""
    resultado = []
    pila = []
    actual = raiz

    while actual is not None or pila:
        # Llegar al nodo más a la izquierda posible
        while actual is not None:
            pila.append(actual)
            actual = actual.izq

        # Procesar el nodo
        actual = pila.pop()
        resultado.append(actual.valor)

        # Pasar al subárbol derecho
        actual = actual.der

    return resultado

print("Inorden iterativo:", inorden_iterativo(arbol.raiz))
```



| Concepto | Idea clave |
|----------|-----------|
| ABB | Izq < raíz < Der |
| Preorden | R-I-D → copiar árbol |
| Inorden | I-R-D → valores ordenados |
| Postorden | I-D-R → eliminar árbol |
| Complejidad recorridos | O(n) tiempo, O(h) espacio |

```python
# Verificar si un árbol binario es un ABB válido

# Estrategia: el inorden de un ABB debe estar ESTRICTAMENTE ordenado
# Complejidad: O(n)

def es_abb(raiz):
    """Retorna True si el árbol cumple la propiedad de ABB"""
    lista = inorden(raiz)
    for i in range(1, len(lista)):
        if lista[i] <= lista[i-1]: # debe ser ESTRICTAMENTE creciente
            return False
    return True


# --- Pruebas ---
# Árbol válido
t1 = Nodo(10)
t1.izq = Nodo(5);  t1.der = Nodo(15)
t1.izq.izq = Nodo(2); t1.izq.der = Nodo(7)
print("¿t1 es ABB?", es_abb(t1))   # True

# Árbol NO válido
t2 = Nodo(10)
t2.izq = Nodo(5);  t2.der = Nodo(15)
t2.izq.izq = Nodo(20)
print("¿t2 es ABB?", es_abb(t2))   # False
```

```python
# Encontrar el k-ésimo elemento menor de un ABB

# Estrategia: hacer inorden (queda ordenado) y tomar el índice k-1.
# Complejidad: O(n) tiempo, O(n) espacio

def k_esimo_menor(raiz, k):
    """Retorna el k-ésimo valor más pequeño (k empieza en 1)."""
    lista = inorden(raiz)
    if k < 1 or k > len(lista):
        return None
    return lista[k-1]

print("1er menor :", k_esimo_menor(arbol.raiz, 1))   # 2
print("3er menor :", k_esimo_menor(arbol.raiz, 3))   # 7
print("5to menor :", k_esimo_menor(arbol.raiz, 5))   # 12
```

```python
arbol = ABB()
for v in [10, 5, 15, 2, 7, 12, 20]:
    arbol.insertar(v)


# Mínimo, Máximo y Altura de un ABB

def minimo(raiz):
    """Mínimo de un ABB: bajamos siempre a la izquierda. O(h)"""
    if raiz is None:
        return None
    while raiz.izq is not None:
        raiz = raiz.izq
    return raiz.valor

def maximo(raiz):
    """Máximo de un ABB: bajamos siempre a la derecha. O(h)"""
    if raiz is None:
        return None
    while raiz.der is not None:
        raiz = raiz.der
    return raiz.valor

def altura(raiz):
    """
    Altura del árbol: número de aristas en el camino más largo.
    Un árbol vacío tiene altura -1; un solo nodo tiene altura 0.
    Complejidad: O(n) — hay que visitar todos los nodos.
    """
    if raiz is None:
        return -1
    return 1 + max(altura(raiz.izq), altura(raiz.der))


print("Mínimo:", minimo(arbol.raiz))      # 2
print("Máximo:", maximo(arbol.raiz))      # 20
print("Altura:", altura(arbol.raiz))      # 2
```

```python
# Contar nodos y calcular la suma de valores

# Ambos son O(n) — recorremos todo el árbol

def contar_nodos(raiz):
    """Cantidad total de nodos del árbol."""
    if raiz is None:
        return 0
    return 1 + contar_nodos(raiz.izq) + contar_nodos(raiz.der)

def suma_valores(raiz):
    """Suma de todos los valores del árbol."""
    if raiz is None:
        return 0
    return raiz.valor + suma_valores(raiz.izq) + suma_valores(raiz.der)

print("Cantidad de nodos:", contar_nodos(arbol.raiz))
print("Suma de valores  :", suma_valores(arbol.raiz))
```

```python
# Reconstruir un árbol dados preorden e inorden

# Idea: el primer elemento del preorden SIEMPRE es la raíz.
#       Buscamos esa raíz en el inorden → todo lo de su izquierda
#       pertenece al subárbol izquierdo, lo de su derecha al derecho
#
# Complejidad: O(n²) en esta versión simple (por el .index())
#              Se puede bajar a O(n) con un diccionario

def construir_arbol(preorden, inorden):
    """Retorna la raíz del árbol reconstruido."""
    if not preorden or not inorden:
        return None

    # La raíz es el primer elemento del preorden
    raiz = Nodo(preorden[0])

    # Buscamos la raíz en el inorden para separar subárboles
    idx = inorden.index(raiz.valor)

    # Subárboles izquierdo y derecho
    inorden_izq  = inorden[:idx]
    inorden_der  = inorden[idx+1:]
    preorden_izq = preorden[1:1+len(inorden_izq)]
    preorden_der = preorden[1+len(inorden_izq):]

    # Recursión
    raiz.izq = construir_arbol(preorden_izq, inorden_izq)
    raiz.der = construir_arbol(preorden_der, inorden_der)

    return raiz


# --- Prueba ---
pre  = [10, 5, 2, 7, 15, 12, 20]
ino  = [2, 5, 7, 10, 12, 15, 20]
nuevo_arbol = construir_arbol(pre, ino)

print("Reconstruido (inorden):", inorden(nuevo_arbol))
print("Reconstruido (preorden):", preorden(nuevo_arbol))
```

```python

# Imprimir valores dentro de un rango [a, b]
# Complejidad: O(k + h) donde k = cantidad de valores en el rango

def rango(raiz, a, b, resultado=None):
    """Retorna todos los valores en el árbol que cumplen a <= x <= b."""
    if resultado is None:
        resultado = []
    if raiz is None:
        return resultado

    # Si el valor actual es mayor que 'a', puede haber valores válidos a la izq
    if raiz.valor > a:
        rango(raiz.izq, a, b, resultado)

    # Si está dentro del rango, lo agregamos
    if a <= raiz.valor <= b:
        resultado.append(raiz.valor)

    # Si el valor actual es menor que 'b', puede haber valores válidos a la der
    if raiz.valor < b:
        rango(raiz.der, a, b, resultado)

    return resultado

print("Valores entre 7 y 15:", rango(arbol.raiz, 7, 15))
```
