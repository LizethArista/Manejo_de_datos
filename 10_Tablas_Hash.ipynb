# ¿Qué es una Tabla Hash?

Una Tabla Hash es una estructura de datos que asocia claves (keys) con valores (values). Su principal ventaja es la velocidad: permite insertar, buscar y eliminar elementos en un tiempo promedio de O(1) (tiempo constante)

- Función Hash: Es el algoritmo que convierte la "clave" en un número entero (índice)

- Arreglo subyacente: El índice calculado indica dónde se guardará el valor en la memoria

- Colisiones: Ocurren cuando dos claves distintas generan el mismo índice. Se resuelven mediante "encadenamiento" (listas en cada casilla) o "direccionamiento abierto" (buscar la siguiente casilla libre

En Python, no necesitamos construir la tabla hash desde cero; usamos diccionarios. Sin embargo, es vital entender su sintaxis y complejidad



image.png

```python
# Creación de un diccionario (Tabla Hash vacía y con datos)

# Ejemplo: Mapeo de entidades federativas con su número de municipios

tabla_municipios = {
    "Aguascalientes": 11,
    "Ciudad de México": 16,
    "Jalisco": 125
}

# Inserción y Actualización (O(1))
tabla_municipios["Nuevo León"] = 51  # Inserta nueva clave-valor
tabla_municipios["Jalisco"] = 126    # Actualiza valor existente

# Búsqueda y Acceso (O(1))
# Forma directa (lanza error si la clave no existe)
municipios_cdmx = tabla_municipios["Ciudad de México"]

# Otra forma usando .get() (retorna None o un valor por defecto si no existe)
municipios_inexistentes = tabla_municipios.get("Zacatecas", 0)

# Eliminación (O(1))
del tabla_municipios["Aguascalientes"]

print("Diccionario final:", tabla_municipios)
```

# Conteo de Frecuencias

En demografía, contar frecuencias es una tarea diaria. Las tablas hash son la estructura óptima para esto


```python
#CONTEO DE FRECUENCIAS MIGATORIAS

# Datos simulados: Estado de origen de mujeres migrantes latinoamericanas
origenes_migrantes = [
    "Honduras", "Colombia", "Honduras", "Venezuela", "Guatemala",
    "Colombia", "Honduras", "El Salvador", "Venezuela", "Colombia"
]

# INICIALIZAMOS LA TABLA HASH (Diccionario)
conteo_paises = {}

# ITERAMOS Y CONTAMOS
for pais in origenes_migrantes:
    # Si el país ya es una clave en el diccionario, sumamos 1
    if pais in conteo_paises:
        conteo_paises[pais] += 1
    # Si no existe, lo inicializamos en 1
    else:
        conteo_paises[pais] = 1

print("Frecuencia de países de origen:", conteo_paises)

# Usando el método .get()
conteo_paises_optimo = {}
for pais in origenes_migrantes:
    # .get(pais, 0) busca el país; si no está, devuelve 0, luego le suma 1
    conteo_paises_optimo[pais] = conteo_paises_optimo.get(pais, 0) + 1

print("Frecuencia (Método óptimo):", conteo_paises_optimo)
```
