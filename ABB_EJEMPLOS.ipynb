# RECORDEMOS QUE...
Un Árbol Binario de Búsqueda (ABB) es una estructura de datos jerárquica en la que cada nodo tiene como máximo dos hijos (izquierdo y derecho)

- Todos los valores en el **subárbol izquierdo** de un nodo son **menores** que el valor del nodo

- Todos los valores en el **subárbol derecho** de un nodo son **mayores** que el valor del nodo


Esta propiedad permite realizar operaciones de búsqueda, inserción y eliminación de manera altamente eficiente, con una complejidad promedio de **O(log n)**, donde *n* es la cantidad de nodos.

## Complejidad Temporal
| Operación   | Caso Promedio | Peor Caso (Árbol desbalanceado) |
|-------------|---------------|---------------------------------|
| Búsqueda    | O(log n)      | O(n)                            |
| Inserción   | O(log n)      | O(n)                            |
| Eliminación | O(log n)      | O(n)                            |
| Recorrido   | O(n)          | O(n)                            |

*El peor caso ocurre cuando el árbol se degenera en una lista enlazada (por ejemplo, al insertar datos ya ordenados).*

```python

# DEFINICIÓN DEL NODO Y DEL ÁRBOL BINARIO DE BÚSQUEDA (ABB)

class Nodo:
    """Representa un nodo individual del árbol binario."""
    def __init__(self, valor):
        self.valor = valor     # Dato almacenado en el nodo
        self.izq = None        # Referencia al hijo izquierdo
        self.der = None        # Referencia al hijo derecho


class ABB:
    """Implementación de un Árbol Binario de Búsqueda"""
    def __init__(self):
        self.raiz = None

    def insertar(self, valor):
        """Inserta un valor en el árbol manteniendo la propiedad del ABB"""
        if self.raiz is None:
            self.raiz = Nodo(valor)
        else:
            self._insertar_rec(self.raiz, valor)

    def _insertar_rec(self, nodo, valor):
        """Método auxiliar recursivo para la inserción"""
        if valor < nodo.valor:
            if nodo.izq is None:
                nodo.izq = Nodo(valor)
            else:
                self._insertar_rec(nodo.izq, valor)
        elif valor > nodo.valor:
            if nodo.der is None:
                nodo.der = Nodo(valor)
            else:
                self._insertar_rec(nodo.der, valor)
        # Si valor == nodo.valor, no se inserta (evita duplicados por defecto)

    def buscar(self, valor):
        """Retorna True si el valor existe en el árbol, False en caso contrario"""
        return self._buscar_rec(self.raiz, valor)

    def _buscar_rec(self, nodo, valor):
        """Método auxiliar recursivo para la búsqueda"""
        if nodo is None:
            return False
        if valor == nodo.valor:
            return True
        if valor < nodo.valor:
            return self._buscar_rec(nodo.izq, valor)
        return self._buscar_rec(nodo.der, valor)
```

## Recorridos en un ABB
Los recorridos permiten visitar todos los nodos del árbol. La diferencia radica en el momento en que se "procesa" la raíz:

1. **Preorden (Raiz → Izquierda → Derecha)**: Útil para copiar o clonar la estructura del árbol

2. **Inorden (Izquierda → Raiz → Derecha)**: En un ABB, este recorrido devuelve los valores **estrictamente ordenados** de menor a mayor

3. **Postorden (Izquierda → Derecha → Raiz)**: Útil para eliminar el árbol (liberar memoria de abajo hacia arriba) o evaluar expresiones matemáticas


```python
# RECORRIDOS Y MÉTODOS AUXILIARES DEL ABB

def preorden(nodo, resultado=None):
    """Preorden: Raiz → Izquierda → Derecha. Complejidad: O(n) tiempo, O(h) espacio."""
    if resultado is None:
        resultado = []
    if nodo is not None:
        resultado.append(nodo.valor)         # 1) Procesar raíz
        preorden(nodo.izq, resultado)        # 2) Recorrer subárbol izquierdo
        preorden(nodo.der, resultado)        # 3) Recorrer subárbol derecho
    return resultado

def inorden(nodo, resultado=None):
    """Inorden: Izquierda → Raiz → Derecha. Devuelve los valores ordenados."""
    if resultado is None:
        resultado = []
    if nodo is not None:
        inorden(nodo.izq, resultado)         # 1) Recorrer subárbol izquierdo
        resultado.append(nodo.valor)         # 2) Procesar raíz
        inorden(nodo.der, resultado)         # 3) Recorrer subárbol derecho
    return resultado

def postorden(nodo, resultado=None):
    """Postorden: Izquierda → Derecha → Raiz. Útil para liberar memoria."""
    if resultado is None:
        resultado = []
    if nodo is not None:
        postorden(nodo.izq, resultado)       # 1) Recorrer subárbol izquierdo
        postorden(nodo.der, resultado)       # 2) Recorrer subárbol derecho
        resultado.append(nodo.valor)         # 3) Procesar raíz
    return resultado

def minimo(raiz):
    """Obtiene el valor mínimo bajando siempre por la izquierda. Complejidad: O(h)."""
    if raiz is None:
        return None
    while raiz.izq is not None:
        raiz = raiz.izq
    return raiz.valor

def maximo(raiz):
    """Obtiene el valor máximo bajando siempre por la derecha. Complejidad: O(h)."""
    if raiz is None:
        return None
    while raiz.der is not None:
        raiz = raiz.der
    return raiz.valor

def altura(raiz):
    """Calcula la altura del árbol (número de aristas en el camino más largo)."""
    if raiz is None:
        return -1
    return 1 + max(altura(raiz.izq), altura(raiz.der))
```

## Gestión de Expedientes Sociodemográficos

 Imaginemos que estamos trabajando con un subconjunto de folios de encuestados del Censo de Población y Vivienda. Necesitamos una estructura que permita registrar nuevos folios y verificar rápidamente si un folio específico ya fue procesado, evitando duplicados y acelerando la búsqueda en comparación con una lista simple


**Ventaja del ABB**: Mientras que buscar en una lista no ordenada toma **O(n)**, en un ABB balanceado toma **O(log n)**, lo cual es util cuando se manejan miles de registros

```python
# SISTEMA DE REGISTRO DE FOLIOS CENSALES

# Inicializamos el árbol para almacenar folios de encuestados
arbol_folios = ABB()

# Simulamos la llegada de folios (se insertan manteniendo el orden lógico)
folios_registrados = [1050, 820, 1200, 750, 900, 1150, 1300]
for folio in folios_registrados:
    arbol_folios.insertar(folio)

print(f"Se han registrado {len(folios_registrados)} folios en el sistema")

# Búsqueda
folio_a_buscar = 900
if arbol_folios.buscar(folio_a_buscar):
    print(f" El folio {folio_a_buscar} YA ha sido procesado")
else:
    print(f" El folio {folio_a_buscar} NO se encuentra en el sistema")

folio_no_existente = 999
if arbol_folios.buscar(folio_no_existente):
    print(f" El folio {folio_no_existente} YA ha sido procesado")
else:
    print(f" El folio {folio_no_existente} NO se encuentra en el sistema (listo para asignar)")

# Obtenemos las estadísticas
print(f"\n Estadísticas del árbol de folios:")
print(f"   - Folio más bajo registrado: {minimo(arbol_folios.raiz)}")
print(f"   - Folio más alto registrado: {maximo(arbol_folios.raiz)}")
print(f"   - Altura del árbol (nivel de anidación): {altura(arbol_folios.raiz)}")
print(f"   - Orden de procesamiento (Inorden): {inorden(arbol_folios.raiz)}")
```

## Estratificación Dinámica y Asignación de Recursos

En el análisis de datos censales o encuestas continuas (como la ENIGH o módulos específicos del INEGI), es frecuente necesitar identificar hogares que caen dentro de un **umbral de vulnerabilidad** específico para asignar recursos o intervenciones sociales

Si los datos llegan en flujo continuo, filtrar un `DataFrame` completo en cada nueva consulta tiene un costo de **O(n)**. Un Árbol Binario de Búsqueda (ABB) reduce este costo a **O(k + h)** para consultas de rango, donde *k* es el número de hogares en el rango y *h* es la altura del árbol


### Características de esta Implementación:

1. **Nodos con Objetos Complejos**: El árbol no solo almacena números, sino diccionarios (o registros) de hogares, ordenados por su `indice_vulnerabilidad`

2. **Consulta de Rango con Agregación**: La función `rango_con_estadisticas` no solo recupera los hogares en el rango, sino que calcula en un solo recorrido la suma de un atributo clave (ej. número de hijas e hijos menores de edad), simulando el cálculo de totales para ponderación

3. **Determinación de Umbrales de Política Pública**: Uso de la función del *k-ésimo elemento menor* para identificar dinámicamente el punto de corte del "10% más vulnerable" sin necesidad de ordenar toda la base de datos desde cero

```python
class Nodo:
    """Nodo del ABB que almacena un registro de hogar completo."""
    def __init__(self, registro):
        self.registro = registro               # Diccionario con datos del hogar
        self.valor_clave = registro['indice_vulnerabilidad'] # Clave de ordenamiento
        self.izq = None
        self.der = None

class ABB_Hogares:
    """Árbol Binario de Búsqueda especializado en registros de hogares"""
    def __init__(self):
        self.raiz = None

    def insertar(self, registro):
        """Inserta un hogar manteniendo el orden por índice de vulnerabilidad"""
        if self.raiz is None:
            self.raiz = Nodo(registro)
        else:
            self._insertar_rec(self.raiz, registro)

    def _insertar_rec(self, nodo, registro):
        valor = registro['indice_vulnerabilidad']
        if valor < nodo.valor_clave:
            if nodo.izq is None:
                nodo.izq = Nodo(registro)
            else:
                self._insertar_rec(nodo.izq, registro)
        elif valor > nodo.valor_clave:
            if nodo.der is None:
                nodo.der = Nodo(registro)
            else:
                self._insertar_rec(nodo.der, registro)
        # Nota: Si el valor es igual, se podría manejar con una lista en el nodo
        # para permitir duplicados, pero por simplicidad se omiten aquí

    def inorden(self, nodo, resultado=None):
        """Devuelve los registros ordenados ascendentemente por vulnerabilidad"""
        if resultado is None:
            resultado = []
        if nodo is not None:
            self.inorden(nodo.izq, resultado)
            resultado.append(nodo.registro)
            self.inorden(nodo.der, resultado)
        return resultado

def rango_con_estadisticas(nodo, min_vuln, max_vuln, resultados=None):
    """
    Recupera hogares en un rango de vulnerabilidad y calcula estadísticas
    agregadas (ej. total de dependientes) en un solo recorrido O(k + h)
    """
    if resultados is None:
        resultados = {'hogares': [], 'total_dependientes': 0, 'conteo': 0}

    if nodo is None:
        return resultados

    # si el valor actual es mayor que el mínimo, buscar a la izquierda
    if nodo.valor_clave > min_vuln:
        rango_con_estadisticas(nodo.izq, min_vuln, max_vuln, resultados)

    # Si está en el rango, se agrega y se acumula la estadística
    if min_vuln <= nodo.valor_clave <= max_vuln:
        resultados['hogares'].append(nodo.registro['folio'])
        resultados['conteo'] += 1
        resultados['total_dependientes'] += nodo.registro.get('dependientes_menores', 0)

    # si el valor actual es menor que el máximo, buscar a la derecha
    if nodo.valor_clave < max_vuln:
        rango_con_estadisticas(nodo.der, min_vuln, max_vuln, resultados)

    return resultados

def k_esimo_menor(nodo, k, contador=None):
    """
    Encuentra el k-ésimo registro con menor vulnerabilidad
    Es util para defiir umbrales de corte para programas sociales (por ejemplo atender al 10% más vulnerable)
    Complejidad: O(h + k)
    """
    if contador is None:
        contador = {'actual': 0, 'resultado': None}

    if nodo is not None and contador['resultado'] is None:
        # Exploramos el subárbol izquierdo
        k_esimo_menor(nodo.izq, k, contador)

        # Procesamos el nodo actual
        if contador['resultado'] is None:
            contador['actual'] += 1
            if contador['actual'] == k:
                contador['resultado'] = nodo.registro

        # Exploramos subárbol derecho solo si aún no se encontró el resultado
        if contador['resultado'] is None:
            k_esimo_menor(nodo.der, k, contador)

    return contador['resultado']
```

```python
# SIMULACIÓN DE ESCENARIO REAL

# Inicializamos el árbol
arbol_vulnerabilidad = ABB_Hogares()

# Simulamos la llegada de registros de hogares (ej. de un flujo de datos del INEGI)
# Cada registro contiene: folio, índice de vulnerabilidad (0-100) y dependientes menores
muestras_hogares = [
    {'folio': 'H001', 'indice_vulnerabilidad': 85, 'dependientes_menores': 3},
    {'folio': 'H002', 'indice_vulnerabilidad': 45, 'dependientes_menores': 0},
    {'folio': 'H003', 'indice_vulnerabilidad': 92, 'dependientes_menores': 4},
    {'folio': 'H004', 'indice_vulnerabilidad': 60, 'dependientes_menores': 1},
    {'folio': 'H005', 'indice_vulnerabilidad': 78, 'dependientes_menores': 2},
    {'folio': 'H006', 'indice_vulnerabilidad': 30, 'dependientes_menores': 0},
    {'folio': 'H007', 'indice_vulnerabilidad': 88, 'dependientes_menores': 3},
]

for hogar in muestras_hogares:
    arbol_vulnerabilidad.insertar(hogar)

print(f" Se han indexado {len(muestras_hogares)} hogares en el ABB")

# CONSULTA 1: Rango con Agregación
# Identificar hogares en "vulnerabilidad alta" (entre 75 y 95) y calcular carga de dependientes
min_rango, max_rango = 75, 95
estadisticas = rango_con_estadisticas(arbol_vulnerabilidad.raiz, min_rango, max_rango)

print(f"\n Análisis de Hogares en Rango de Vulnerabilidad [{min_rango} - {max_rango}]:")
print(f"   - Folios identificados: {estadisticas['hogares']}")
print(f"   - Total de hogares en este rango: {estadisticas['conteo']}")
print(f"   - Total de dependientes menores a atender: {estadisticas['total_dependientes']}")
if estadisticas['conteo'] > 0:
    promedio_dep = estadisticas['total_dependientes'] / estadisticas['conteo']
    print(f"   - Promedio de dependientes por hogar en este rango: {promedio_dep:.2f}")

# CONSULTA 2: Determinación de Umbral de Política Pública

# Supongamos que el presupuesto solo alcanza para atender a los 3 hogares más vulnerables
# Necesitamos saber cuál es el índice de vulnerabilidad de corte (el 3er menor en la cola de alta vulnerabilidad,
# o equivalentemente, el 3er más alto. Para usar la función 'k_esimo_menor', ordenamos ascendentemente)
# Si hay 7 hogares, el 3er más vulnerable es el (7 - 3 + 1) = 5to elemento en orden ascendente

k = 5 # 5to elemento en orden ascendente (equivalente al 3er más vulnerable)
hogar_umbral = k_esimo_menor(arbol_vulnerabilidad.raiz, k)

print(f"\n Definición de Umbral para Programa Social:")
print(f"   - El hogar que marca el límite de atención (k={k} en orden asc. de vulnerabilidad) es:")
print(f"     Folio: {hogar_umbral['folio']}, Índice: {hogar_umbral['indice_vulnerabilidad']}")
print(f"   - Esto significa que todos los hogares con índice >= {hogar_umbral['indice_vulnerabilidad']} deben ser priorizados.")

# Verificación del ordenamiento global (Inorden)
print(f"\n Verificación del ordenamiento global (Inorden):")
registros_ordenados = arbol_vulnerabilidad.inorden(arbol_vulnerabilidad.raiz)
for r in registros_ordenados:
    print(f"   Folio: {r['folio']} | Vulnerabilidad: {r['indice_vulnerabilidad']} | Dependientes: {r['dependientes_menores']}")
```
