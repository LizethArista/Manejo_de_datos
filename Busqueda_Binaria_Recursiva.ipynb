## ¿Qué es la Búsqueda Binaria?

La **búsqueda binaria** es un algoritmo de búsqueda extremadamente eficiente que funciona **SOLO en listas ordenadas**

Imagina que buscas la palabra "Python" en un diccionario físico:

1. **Abres el diccionario por la mitad**
2. Ves una palabra que empieza con "M"
3. Como "Python" viene después de "M", **descartas la mitad izquierda**
4. Abres la mitad derecha por la mitad
5. Ves una palabra que empieza con "R"
6. Como "Python" viene después de "R", **descartas la mitad izquierda**
7. Repites hasta encontrar la palabra

**En cada paso, reduces el problema a la MITAD.**


Lista ordenada: `[2, 5, 8, 12, 16, 23, 38, 56, 72, 91]`
Buscamos: `23`



Paso 1: Lista completa [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]

Medio = índice 4 → valor 16

23 > 16 → descartamos la mitad izquierda

Paso 2: Sublista [23, 38, 56, 72, 91]

Medio = índice 7 → valor 56

23 < 56 → descartamos la mitad derecha

Paso 3: Sublista [23, 38]

Medio = índice 5 → valor 23

¡ENCONTRADO!


Solo 3 pasos para buscar en 10 elementos

- **Mejor caso:** O(1) — el elemento está en el medio
- **Peor caso:** O(log n) — dividimos hasta llegar a 1 elemento
- **Requisito:** La lista DEBE estar ordenada

En cada paso, **dividimos el problema a la mitad**:

| Paso | Elementos restantes |
|------|---------------------|
| 0    | n                   |
| 1    | n/2                 |
| 2    | n/4                 |
| 3    | n/8                 |
| ...  | ...                 |
| k    | n/2^k = 1           |

Cuando llegamos a 1 elemento:

n/2^k = 1

n = 2^k

k = log₂(n)



### Ejemplos

| Tamaño de lista (n) | Pasos máximos (log₂ n) |
|---------------------|------------------------|
| 10                  | ~3.3 → 4 pasos         |
| 100                 | ~6.6 → 7 pasos         |
| 1,000               | ~10 → 10 pasos         |
| 1,000,000           | ~20 → 20 pasos         |
| 1,000,000,000       | ~30 → 30 pasos         |
| 4,000,000,000       | ~32 → 32 pasos         |

- **Búsqueda lineal O(n):** hasta 4,000,000,000 pasos (horas)
- **Búsqueda binaria O(log n):** máximo 32 pasos (microsegundos)

```python

# BÚSQUEDA BINARIA ITERATIVA

def busqueda_binaria_iterativa(lista_ordenada, objetivo):
    """
    Busca 'objetivo' en una lista ordenada usando un bucle.

    Complejidad:
        Tiempo: O(log n)
        Espacio: O(1) - no usa memoria extra

    Args:
        lista_ordenada: Lista ordenada de menor a mayor
        objetivo: Valor a buscar

    Returns:
        Índice del elemento si se encuentra, -1 si no existe
    """
    # Definimos los límites de búsqueda
    izquierda = 0
    derecha = len(lista_ordenada) - 1

    # Mientras el rango de búsqueda sea válido
    while izquierda <= derecha:
        # Calculamos el punto medio (evitando overflow)
        medio = (izquierda + derecha) // 2

        # Caso 1: Encontramos el objetivo
        if lista_ordenada[medio] == objetivo:
            return medio  # Retornamos el índice

        # Caso 2: El objetivo es mayor → buscamos en la mitad derecha
        elif lista_ordenada[medio] < objetivo:
            izquierda = medio + 1

        # Caso 3: El objetivo es menor → buscamos en la mitad izquierda
        else:
            derecha = medio - 1

    # Si salimos del bucle, el elemento no existe
    return -1


print("=== BÚSQUEDA BINARIA ITERATIVA ===\n")

# Lista ordenada de ejemplo
lista = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(f"Lista: {lista}\n")

# Prueba 1: Elemento existe
objetivo = 23
posicion = busqueda_binaria_iterativa(lista, objetivo)
print(f"Buscar {objetivo}: posición {posicion}")

# Prueba 2: Elemento no existe
objetivo = 25
posicion = busqueda_binaria_iterativa(lista, objetivo)
print(f"Buscar {objetivo}: posición {posicion} (no existe)")

# Prueba 3: Primer elemento
objetivo = 2
posicion = busqueda_binaria_iterativa(lista, objetivo)
print(f"Buscar {objetivo}: posición {posicion}")

# Prueba 4: Último elemento
objetivo = 91
posicion = busqueda_binaria_iterativa(lista, objetivo)
print(f"Buscar {objetivo}: posición {posicion}")
```

## Búsqueda Binaria Recursiva

La búsqueda binaria es **naturalmente recursiva** porque en cada paso:
1. Dividimos el problema a la mitad
2. Resolvemos el subproblema (llamada recursiva)
3. Combinamos el resultado

Para buscar en una lista entre los índices `[izquierda, derecha]`:

1. **Caso base:** Si `izquierda > derecha`, el elemento no existe → retornar -1
2. **Paso recursivo:**
   - Calcular el medio
   - Si el medio es el objetivo → retornar el índice
   - Si el objetivo es mayor → buscar recursivamente en `[medio+1, derecha]`
   - Si el objetivo es menor → buscar recursivamente en `[izquierda, medio-1]`

busqueda([2, 5, 8, 12, 16, 23, 38, 56, 72, 91], 23, 0, 9)


│

├─► medio = 4, valor = 16

├─► 23 > 16 → buscar en [5, 9]

│

└─► busqueda([..., 23, 38, 56, 72, 91], 23, 5, 9)

│

├─► medio = 7, valor = 56

├─► 23 < 56 → buscar en [5, 6]

│

└─► busqueda([..., 23, 38], 23, 5, 6)

│

├─► medio = 5, valor = 23

├─► ¡ENCONTRADO! → retornar 5



```python

# BÚSQUEDA BINARIA RECURSIVA


def busqueda_binaria_recursiva(lista_ordenada, objetivo, izquierda=None, derecha=None):
    """
    Busca 'objetivo' en una lista ordenada usando recursión.

    ESTRATEGIA RECURSIVA:
    1. Caso base: si izquierda > derecha, el elemento no existe
    2. Calcular el punto medio
    3. Si el medio es el objetivo, retornar el índice
    4. Si el objetivo es mayor, buscar recursivamente en la mitad derecha
    5. Si el objetivo es menor, buscar recursivamente en la mitad izquierda

    Complejidad:
        Tiempo: O(log n)
        Espacio: O(log n) - pila de recursión

    Args:
        lista_ordenada: Lista ordenada de menor a mayor
        objetivo: Valor a buscar
        izquierda: Índice inicial del rango de búsqueda (por defecto 0)
        derecha: Índice final del rango de búsqueda (por defecto len(lista)-1)

    Returns:
        Índice del elemento si se encuentra, -1 si no existe
    """
    # INICIALIZACIÓN: Si es la primera llamada, configurar rangos

    if izquierda is None:
        izquierda = 0
    if derecha is None:
        derecha = len(lista_ordenada) - 1


    # CASO BASE 1: Rango inválido, el elemento no existe

    if izquierda > derecha:
        return -1  # No encontramos el elemento

    # PASO RECURSIVO: Calcular el punto medio

    medio = (izquierda + derecha) // 2

    # CASO BASE 2: Encontramos el objetivo

    if lista_ordenada[medio] == objetivo:
        return medio  # ¡Encontrado! Retornamos el índice


    # PASO RECURSIVO 1: Objetivo es mayor, se busca en mitad derecha

    elif lista_ordenada[medio] < objetivo:
        # Llamada recursiva con nuevo rango [medio+1, derecha]
        return busqueda_binaria_recursiva(
            lista_ordenada,
            objetivo,
            medio + 1,  # Nuevo límite izquierdo
            derecha     # Límite derecho se mantiene
        )


    # PASO RECURSIVO 2: Objetivo es menor, se busca en mitad izquierda

    else:
        # Llamada recursiva con nuevo rango [izquierda, medio-1]
        return busqueda_binaria_recursiva(
            lista_ordenada,
            objetivo,
            izquierda,  # Límite izquierdo se mantiene
            medio - 1   # Nuevo límite derecho
        )

print("=== BÚSQUEDA BINARIA RECURSIVA ===\n")

# Lista ordenada de ejemplo
lista = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(f"Lista: {lista}\n")

# Prueba 1: Elemento existe
objetivo = 23
posicion = busqueda_binaria_recursiva(lista, objetivo)
print(f"Buscar {objetivo}: posición {posicion}")

```

```python
# BÚSQUEDA BINARIA RECURSIVA

def busqueda_binaria_con_trazas(lista_ordenada, objetivo, izquierda=None, derecha=None, profundidad=0):
    """
    Búsqueda binaria recursiva que muestra cada paso del proceso
    Útil para visualizar el árbol de recursión
    """
    # Inicialización
    if izquierda is None:
        izquierda = 0
    if derecha is None:
        derecha = len(lista_ordenada) - 1

    # Indentación para ver la profundiddad
    indent = "  " * profundidad

    # Muestro el rango actual de búsqueda
    print(f"{indent} Rango: [{izquierda}, {derecha}]")
    print(f"{indent}   Sublista: {lista_ordenada[izquierda:derecha+1]}")

    # CASO BASE 1: Rango inválido
    if izquierda > derecha:
        print(f"{indent}    Rango vacío, elemento no existe\n")
        return -1

    # Calculo el punto medio
    medio = (izquierda + derecha) // 2
    valor_medio = lista_ordenada[medio]

    print(f"{indent}    Medio = índice {medio}, valor = {valor_medio}")

    # CASO BASE 2: Encontramos el objetivo
    if valor_medio == objetivo:
        print(f"{indent}    ¡ENCONTRADO! {objetivo} está en posición {medio}\n")
        return medio

    # PASO RECURSIVO 1: Buscar en la mitad derecha
    elif valor_medio < objetivo:
        print(f"{indent}   → {objetivo} > {valor_medio} → buscar en MITAD DERECHA\n")
        return busqueda_binaria_con_trazas(
            lista_ordenada, objetivo, medio + 1, derecha, profundidad + 1
        )

    # PASO RECURSIVO 2: Buscar en la mitad izquierda
    else:
        print(f"{indent}   → {objetivo} < {valor_medio} → buscar en MITAD IZQUIERDA\n")
        return busqueda_binaria_con_trazas(
            lista_ordenada, objetivo, izquierda, medio - 1, profundidad + 1
        )

print("=== TRAZANDO BÚSQUEDA BINARIA RECURSIVA ===\n")

lista = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
objetivo = 23

print(f"Lista: {lista}")
print(f"Buscando: {objetivo}\n")
print("="*60)

resultado = busqueda_binaria_con_trazas(lista, objetivo)

print("="*60)
print(f"\n Resultado final: posición {resultado}")
print(f"   Valor en esa posición: {lista[resultado]}")
```

```python
# COMPARACIÓN: ITERATIVO vs RECURSIVO

def medir_tiempo(funcion, *args, repeticiones=100):
    """Mide el tiempo promedio de ejecutar una función."""
    tiempos = []
    for _ in range(repeticiones):
        inicio = time.time()
        funcion(*args)
        fin = time.time()
        tiempos.append(fin - inicio)
    return sum(tiempos) / len(tiempos)

# Generamos una lista grande ordenada
n = 1_000_000
lista_grande = list(range(n))

# Elementos a buscar (diferentes posiciones)
objetivos = [
    n // 4,        # Primer cuarto
    n // 2,        # Medio
    3 * n // 4,    # Tercer cuarto
    n - 1,         # Último elemento
    n + 100        # No existe
]

print("=== COMPARACIÓN: ITERATIVO vs RECURSIVO ===\n")
print(f"Lista de {n:,} elementos ordenados\n")
print(f"{'Objetivo':>12} | {'Iterativo ':>15} | {'Recursivo ':>15} ")
print("-" * 65)

for objetivo in objetivos:
    # Medir iterativo
    t_iter = medir_tiempo(busqueda_binaria_iterativa, lista_grande, objetivo)

    # Medir recursivo
    t_rec = medir_tiempo(busqueda_binaria_recursiva, lista_grande, objetivo)


    print(f"{objetivo:>12,} | {t_iter*1_000_000:>15.3f} | {t_rec*1_000_000:>15.3f}")
```

```python
# COMPARACIÓN: BINARIA vs LINEAL

def busqueda_lineal(lista, objetivo):
    """Búsqueda lineal O(n)."""
    for i in range(len(lista)):
        if lista[i] == objetivo:
            return i
    return -1

# Diferentes tamaños de lista
tamanos = [100, 1000, 10000, 100000, 1000000]

print("=== BÚSQUEDA LINEAL vs BINARIA ===\n")
print(f"{'n':>10} | {'Lineal ':>12} | {'Binaria':>12}")
print("-" * 55)

for n in tamanos:
    lista = list(range(n))
    objetivo = n - 1  # Peor caso: último elemento

    # Medir lineal
    t_lin = medir_tiempo(busqueda_lineal, lista, objetivo, repeticiones=10)

    # Medir binaria
    t_bin = medir_tiempo(busqueda_binaria_iterativa, lista, objetivo, repeticiones=100)

    # Calcular ratio
    ratio = t_lin / t_bin if t_bin > 0 else 0

    print(f"{n:>10,} | {t_lin*1_000_000:>12.3f} | {t_bin*1_000_000:>12.3f}")

print("\n Observaciones:")
print("• La búsqueda binaria es MUCHO más rápida")
print("• La diferencia crece con el tamaño de la lista")
print("• Con 1 millón de elementos, binaria es ~100,000 veces más rápida")
print("• Por eso la búsqueda binaria es esencial en bases de datos")
```

```python
# En matemáticas, para encontrar la raíz de una función monótona:

def buscar_raiz(f, objetivo, izquierda, derecha, tolerancia=1e-6):
    """Busca x tal que f(x) = objetivo usando búsqueda binaria."""
    while derecha - izquierda > tolerancia:
        medio = (izquierda + derecha) / 2
        if f(medio) < objetivo:
            izquierda = medio
        else:
            derecha = medio
    return (izquierda + derecha) / 2
```

```python
# --- Ejemplo 1: Encontrar la primera ocurrencia ---
def busqueda_primera_ocurrencia(lista_ordenada, objetivo):
    """
    Encuentra la PRIMERA posición del objetivo en una lista con duplicados.
    """
    izquierda = 0
    derecha = len(lista_ordenada) - 1
    resultado = -1

    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2

        if lista_ordenada[medio] == objetivo:
            resultado = medio  # Guardamos la posición
            derecha = medio - 1  # Pero seguimos buscando a la izquierda
        elif lista_ordenada[medio] < objetivo:
            izquierda = medio + 1
        else:
            derecha = medio - 1

    return resultado

# Prueba
lista_con_duplicados = [1, 2, 2, 2, 2, 3, 4, 5]
print("Ejemplo 1: Primera ocurrencia")
print(f"Lista: {lista_con_duplicados}")
print(f"Primera posición de 2: {busqueda_primera_ocurrencia(lista_con_duplicados, 2)}")
print(f"Primera posición de 3: {busqueda_primera_ocurrencia(lista_con_duplicados, 3)}\n")

```

```python
# --- Ejemplo 2: Encontrar la última ocurrencia ---
def busqueda_ultima_ocurrencia(lista_ordenada, objetivo):
    """
    Encuentra la ÚLTIMA posición del objetivo en una lista con duplicados.
    """
    izquierda = 0
    derecha = len(lista_ordenada) - 1
    resultado = -1

    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2

        if lista_ordenada[medio] == objetivo:
            resultado = medio  # Guardamos la posición
            izquierda = medio + 1  # Pero seguimos buscando a la derecha
        elif lista_ordenada[medio] < objetivo:
            izquierda = medio + 1
        else:
            derecha = medio - 1

    return resultado

print("Ejemplo 2: Última ocurrencia")
print(f"Lista: {lista_con_duplicados}")
print(f"Última posición de 2: {busqueda_ultima_ocurrencia(lista_con_duplicados, 2)}")
print(f"Última posición de 3: {busqueda_ultima_ocurrencia(lista_con_duplicados, 3)}\n")

```

```python
# --- Ejemplo 3: Encontrar el elemento más cercano ---
def busqueda_mas_cercano(lista_ordenada, objetivo):
    """
    Encuentra el elemento más cercano al objetivo.
    """
    izquierda = 0
    derecha = len(lista_ordenada) - 1

    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2

        if lista_ordenada[medio] == objetivo:
            return medio

        # Calculamos distancias
        dist_medio = abs(lista_ordenada[medio] - objetivo)

        # Verificamos vecinos
        if medio > 0:
            dist_izq = abs(lista_ordenada[medio - 1] - objetivo)
            if dist_izq < dist_medio:
                derecha = medio - 1
                continue

        if medio < len(lista_ordenada) - 1:
            dist_der = abs(lista_ordenada[medio + 1] - objetivo)
            if dist_der < dist_medio:
                izquierda = medio + 1
                continue

        return medio

    return -1

lista_ejemplo = [1, 3, 5, 7, 9, 11, 13]
print("Ejemplo 3: Elemento más cercano")
print(f"Lista: {lista_ejemplo}")
print(f"Más cercano a 6: índice {busqueda_mas_cercano(lista_ejemplo, 6)} (valor {lista_ejemplo[busqueda_mas_cercano(lista_ejemplo, 6)]})")
print(f"Más cercano a 8: índice {busqueda_mas_cercano(lista_ejemplo, 8)} (valor {lista_ejemplo[busqueda_mas_cercano(lista_ejemplo, 8)]})\n")
```

```python
# --- Ejemplo 4: Buscar en matriz ordenada ---
def busqueda_en_matriz(matriz, objetivo):
    """
    Busca un elemento en una matriz donde cada fila y columna están ordenadas
    """
    if not matriz or not matriz[0]:
        return -1, -1

    filas = len(matriz)
    columnas = len(matriz[0])

    # Empezamos desde la esquina superior derecha
    i, j = 0, columnas - 1

    while i < filas and j >= 0:
        if matriz[i][j] == objetivo:
            return i, j
        elif matriz[i][j] > objetivo:
            j -= 1  # Moverse a la izquierda
        else:
            i += 1  # Moverse hacia abajo

    return -1, -1

matriz_ejemplo = [
    [1, 4, 7, 11],
    [2, 5, 8, 12],
    [3, 6, 9, 16],
    [10, 13, 14, 17]
]

print("Ejemplo 4: Búsqueda en matriz ordenada")
print("Matriz:")
for fila in matriz_ejemplo:
    print(f"  {fila}")
pos = busqueda_en_matriz(matriz_ejemplo, 8)
print(f"Buscar 8: posición {pos}")
pos = busqueda_en_matriz(matriz_ejemplo, 14)
print(f"Buscar 14: posición {pos}")
```
