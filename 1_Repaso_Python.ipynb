## 5. Ciclo `while`

El ciclo `while` repite instrucciones mientras una condición sea verdadera

Es importante actualizar la condición para evitar ciclos infinitos

````python
contador = 1

while contador <= 5:
    print("Contador:", contador)
    contador += 1
````

````python
numero = 0

while True:
    numero += 1

    if numero == 3:
        continue

    if numero > 5:
        break #termina el ciclo

    print("Número:", numero)
````

## 7. Estructuras de datos

Repasaremos tres estructuras muy usadas:

| Estructura | Se declara con | ¿Es mutable? |
|---|---|---|
| Lista | `[]` | Sí |
| Tupla | `()` | No |
| Diccionario | `{}` | Sí |

````python
# Lista
frutas = ["manzana", "pera", "mango"]
print(frutas)

frutas.append("uva")
print(frutas)

frutas[0] = "manzana verde"
print(frutas)

print("Primera fruta:", frutas[0])
print("Última fruta:", frutas[-1])

# Tupla
coordenada = (10, 20)
print(coordenada)
print("Primer valor:", coordenada[0])

# Diccionario
persona = {
    "nombre": "Luis",
    "edad": 22,
    "materias": ["POO", "Datos"]
}

print(persona)
print(persona["nombre"])
print(persona.get("promedio", 0))

persona["edad"] = 23
print(persona)

for clave, valor in persona.items():
    print(clave, "->", valor)
````

## 8. Objetos, estado y servicios

En programación orientada a objetos:

- Un **objeto** representa una entidad del problema
- El **estado** son sus atributos o datos
- Los **servicios** son sus métodos, es decir, las acciones que puede realizar

Por ejemplo...

Una mascota puede tener:

- Estado: nombre, energía
- Servicios: jugar, dormir

````python
class Mascota:
    def __init__(self, nombre, energia=100):
        self.nombre = nombre
        self.energia = energia

    def jugar(self):
        if self.energia >= 20:
            self.energia -= 20
            return f"{self.nombre} jugó. Energía actual: {self.energia}"
        else:
            return f"{self.nombre} está cansado. Energía actual: {self.energia}"

    def dormir(self):
        self.energia += 30
        return f"{self.nombre} durmió. Energía actual: {self.energia}"


mi_mascota = Mascota("Firulais", 50)

print(mi_mascota.jugar())
print(mi_mascota.dormir())
print(mi_mascota.jugar())
````

__init__ es el constructor

self hace referencia al objeto actual

nombre y energia son parte del estado

jugar() y dormir() son servicios o métodos

## 9. Clases y métodos

Una clase es una plantilla para crear objetos

Por ejemplo, una cuenta bancaria puede tener:

- Estado: titular, saldo
- Métodos: depositar, retirar, consultar

````python
class Cuenta:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, monto):
        if monto <= 0:
            return "Monto inválido"
        self.saldo += monto
        return f"Depósito de {monto}. Saldo actual: {self.saldo}"

    def retirar(self, monto):
        if monto <= 0 or monto > self.saldo:
            return "Retiro inválido"
        self.saldo -= monto
        return f"Retiro de {monto}. Saldo actual: {self.saldo}"

    def consultar(self):
        return f"{self.titular} tiene {self.saldo}"


cuenta1 = Cuenta("Ana", 100)

print(cuenta1.consultar())
print(cuenta1.depositar(50))
print(cuenta1.retirar(30))
print(cuenta1.consultar())
````

## 10. Herencia

La herencia permite crear una clase nueva a partir de una existente.

La clase nueva puede:

- Reutilizar métodos
- Agregar atributos nuevos
- Agregar métodos nuevos
- Modificar métodos existentes

````python
class CuentaAhorro(Cuenta): #hereda de Cuenta
    def __init__(self, titular, saldo=0, meta=1000):
        super().__init__(titular, saldo)  #llama al constructor de la clase padre
        self.meta = meta #agrega el atributo meta

    def avanzar_meta(self, monto): #agrega el método avanzar_meta

        self.depositar(monto)

        if self.saldo >= self.meta:
            return f"¡Meta alcanzada! Saldo: {self.saldo}, Meta: {self.meta}"
        else:
            return f"Avance: {self.saldo} de {self.meta}"


ahorro = CuentaAhorro("Beto", 900, 1000)

print(ahorro.consultar())
print(ahorro.avanzar_meta(50))
print(ahorro.retirar(20))
print(ahorro.avanzar_meta(10))

print(isinstance(ahorro, CuentaAhorro))
print(isinstance(ahorro, Cuenta))
````
