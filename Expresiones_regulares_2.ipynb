# EXPRESIONES REGULARES



````python
import re

print("=" * 60)
print("REPASO RAPIDO DE METODOS")
print("=" * 60)

texto = "Orden #12345: Producto XYZ - Precio: $99.99"
print("Texto:", texto)
print("=" * 60)

# match - solo al inicio
print("\n1. re.match() - Solo al inicio:")
print(f"   re.match(r'Orden', texto): {re.match(r'Orden', texto)}")

# search - en cualquier parte
print("\n2. re.search() - En cualquier parte:")
print(f"   re.search(r'\\d+', texto).group(): {re.search(r'\d+', texto).group()}")

# fullmatch - cadena completa
print("\n3. re.fullmatch() - Cadena completa:")
print(f"   re.fullmatch(r'Orden.*', texto): {re.fullmatch(r'Orden.*', texto)}")

# findall - todas las coincidencias
print("\n4. re.findall() - Todas las coincidencias:")
print(f"   re.findall(r'\\d+', texto): {re.findall(r'\d+', texto)}")

# sub - reemplazar
print("\n5. re.sub() - Reemplazar:")
print(f"   re.sub(r'\\$\\d+\\.\\d+', '$XX.XX', texto): {re.sub(r'\$\d+\.\d+', '$XX.XX', texto)}")

# split - dividir
print("\n6. re.split() - Dividir:")
print(f"   re.split(r'[:\\-]', texto): {re.split(r'[:\\-]', texto)}")
````

## Patrones Avanzados: Lookahead y Lookbehind

Permiten buscar patrones **sin consumir caracteres**. Es decir, verifican una condicion pero no avanzan el cursor.

### Tipos:

| Tipo | Sintaxis | Significado |
|------|----------|-------------|
| Lookahead positivo | `(?=...)` | Coincide si lo que SIGUE cumple |
| Lookahead negativo | `(?!...)` | Coincide si lo que SIGUE NO cumple |
| Lookbehind positivo | `(?<=...)` | Coincide si lo que PRECEDE cumple |
| Lookbehind negativo | `(?<!...)` | Coincide si lo que PRECEDE NO cumple |

### Analogia:
Imagina que lees un libro. El lookahead es como mirar la siguiente palabra sin mover tu dedo. El lookbehind es como mirar la palabra anterior sin retroceder.

````python

# Lookahead positivo (?=...)

print(" Lookahead positivo (?=...):")
texto = "precio100 precio200 300precio"
patron = r'\d+(?=precio)'
print(f"   Texto: {texto}")
print(f"   Patron: {patron}")
print(f"   Resultado: {re.findall(patron, texto)}")  # ['100', '200']
# Busca digitos SOLO si despues viene "precio"

````

````python
# Lookahead negativo (?!...)

print(" Lookahead negativo (?!...):")
patron = r'\d+(?!precio)'
print(f"   Patron: {patron}")
print(f"   Resultado: {re.findall(patron, texto)}")  # ['100', '200', '300']
# Explicacion: busca digitos SOLO si despues NO viene "precio"

````

````python
# Lookbehind positivo (?<=...)

print(" Lookbehind positivo (?<=...):")
texto = "$100 $200 E300"
patron = r'(?<=\$)\d+'
print(f"   Texto: {texto}")
print(f"   Patron: {patron}")
print(f"   Resultado: {re.findall(patron, texto)}")  # ['100', '200']
# Explicacion: busca digitos SOLO si antes viene "$"

````

````python
# Lookbehind negativo (?<!...)

print(" Lookbehind negativo (?<!...):")
patron = r'(?<!\$)\d+'
print(f"   Patron: {patron}")
print(f"   Resultado: {re.findall(patron, texto)}")  # ['300']
# Explicacion: busca digitos SOLO si antes NO viene "$"
````

## El poder de `re.split()` con Lookahead

### El problema:
Imagina que tienes la expresion `"-1/2+1/3-1/6"` y quieres dividirla en fracciones individuales: `["-1/2", "+1/3", "-1/6"]`.

### El error comun:
```python
re.split(r'[+-]', "-1/2+1/3-1/6")
# Resultado: ['', '1/2', '1/3', '1/6']
# PROBLEMA: perdiste los signos + y - !

````python
re.split(r'[+-]', "1/6-1/2+1/3-1/6")
````

````python
re.split(r'(?=[+-])', "1/6 -1/2+1/3-1/6")
# Resultado: ['1/6 ', '-1/2', '+1/3', '-1/6']
# Los signos se CONSERVAN porque el lookahead no los consume
````

Como funciona:

El patron (?=[+-]) le dice a Python:

"Divide justo ANTES de que aparezca un + o un -, pero NO te los comas ni los borres"

El lookahead verifica la condicion sin avanzar el cursor, por lo que el delimitador permanece en el resultado.


### Casos de uso reales:

- Parsear expresiones matematicas (como el problema #592)

- Dividir CSVs manteniendo los separadores

- Separar palabras compuestas por guiones

````python
expresion = "-1/2+1/3-1/6"
print(f"\nExpresion original: '{expresion}'")

# Enfoque INCORRECTO: split sin lookahead

print("\n--- Enfoque INCORRECTO: re.split(r'[+-]', ...) ---")
partes_mal = re.split(r'[+-]', expresion)
print(f"   Resultado: {partes_mal}")
print("   Problema: perdiste los signos + y -")
print("   Las fracciones quedan invalidas: '1/2' en vez de '-1/2'")

````

````python
# Enfoque CORRECTO: split con lookahead

print("\n--- Enfoque CORRECTO: re.split(r'(?=[+-])', ...) ---")
partes_bien = re.split(r'(?=[+-])', expresion)
print(f"   Resultado: {partes_bien}")
print("   Los signos se conservan porque el lookahead no los consume")

# Limpieza de strings vacios (puede aparecer uno al inicio)
partes_bien = [p for p in partes_bien if p]
print(f"   Despues de limpiar vacios: {partes_bien}")

````

````python
# Otro ejemplo: dividir un CSV manteniendo comas

csv = "nombre,edad,ciudad"
print(f"\nCSV original: '{csv}'")

print("\nSin lookahead:")
print(f"   re.split(r',', csv): {re.split(r',', csv)}")

print("\nCon lookahead:")
print(f"   re.split(r'(?=,)', csv): {re.split(r'(?=,)', csv)}")
````

## Modificadores (Flags)

Los flags modifican el comportamiento del motor regex:

| Flag | Alternativa | Efecto |
|------|-------------|--------|
| `re.IGNORECASE` | `re.I` | No distingue mayusculas/minusculas |
| `re.MULTILINE` | `re.M` | `^` y `$` funcionan por linea |
| `re.DOTALL` | `re.S` | `.` incluye saltos de linea |

### Como usarlos:
```python
re.findall(patron, texto, re.IGNORECASE | re.MULTILINE)


Puedes combinar varios flags con el operador | (OR bitwise)


````python

texto = "Hola\nMUNDO\nhola"
print(f"\nTexto:\n{texto}")

# re.IGNORECASE
print("re.IGNORECASE - No distingue mayusculas:")
print(f"   re.findall(r'hola', texto, re.IGNORECASE): {re.findall(r'hola', texto, re.IGNORECASE)}")
````

````python
# re.MULTILINE

print("re.MULTILINE - ^ y $ funcionan por linea:")
resultado = re.findall(r'^hola', texto, re.MULTILINE | re.IGNORECASE)
print(f"   re.findall(r'^hola', texto, re.MULTILINE | re.IGNORECASE): {resultado}")

````

````python
# re.DOTALL

print(" re.DOTALL - . incluye saltos de linea:")
resultado = re.findall(r'Hola.*mundo', texto, re.DOTALL | re.IGNORECASE)
print(f"   re.findall(r'Hola.*mundo', texto, re.DOTALL | re.IGNORECASE): {resultado}")
````
