# QuickSort

- **Pivote:** Elemento de referencia para particionar la lista
- **Particionamiento:** Dividir la lista en menores y mayores al pivote
- **Recursión:** Aplicar el algoritmo a las sublists resultantes
- **Complejidad:** O(n log n) promedio, O(n²) peor caso

```python
import time
import random
import matplotlib.pyplot as plt
import numpy as np
```

## ¿Qué es QuickSort?

QuickSort es un algoritmo de ordenamiento basado en la estrategia **"Divide y Vencerás"**

1. **Elegir un pivote:** Seleccionamos un elemento de la lista (puede ser el primero, el último, el del medio, o uno aleatorio)
2. **Particionar:** Reorganizamos la lista de modo que:
   - Todos los elementos **menores** que el pivote queden a su **izquierda**
   - Todos los elementos **mayores** que el pivote queden a su **derecha**
   - El pivote queda en su **posición final correcta**
3. **Recursión:** Aplicamos el mismo proceso a las dos sublistas resultantes (izquierda y derecha)


Lista: `[8, 3, 1, 7, 0, 10, 2]`

Elegimos pivote = 2 (último elemento)

**Particionamiento:**
- Menores que 2: `[0, 1]`
- Pivote: `2`
- Mayores que 2: `[8, 3, 7, 10]`

Resultado después de particionar: `[0, 1, 2, 8, 3, 7, 10]`

Ahora el 2 está en su posición final. Aplicamos recursión a:
- Izquierda: `[0, 1]` (ya ordenado)
- Derecha: `[8, 3, 7, 10]` (requiere más trabajo)

### Complejidad
- **Mejor caso:** O(n log n) - particiones balanceadas
- **Peor caso:** O(n²) - particiones muy desbalanceadas
- **Promedio:** O(n log n) - con alta probabilidad
- **Espacio:** O(log n) - por la pila de recursión
- **Estable:** No - puede cambiar el orden relativo de elementos iguales

## El Particionamiento

El particionamiento es la operación clave de QuickSort. Existen varios esquemas:

### Esquema de Lomuto (Más simple de entender)

1. Elegimos el **último elemento** como pivote
2. Mantenemos un puntero `i` que marca la frontera entre elementos menores y mayores
3. Recorremos con un puntero `j` desde el inicio hasta el penúltimo elemento
4. Si `lista[j] < pivote`, intercambiamos `lista[i]` con `lista[j]` y avanzamos `i`
5. Finalmente, colocamos el pivote en la posición `i`

### Esquema de Hoare (Más eficiente)

1. Usamos dos punteros, uno desde la izquierda y otro desde la derecha
2. Avanzan hacia el centro intercambiando elementos cuando es necesario
3. Más eficiente en la práctica (menos intercambios)

### Estrategias de Selección del Pivote

| Estrategia | Ventaja | Desventaja |
|------------|---------|------------|
| **Primer elemento** | Simple | Malo si la lista ya está ordenada |
| **Último elemento** | Simple | Malo si la lista ya está ordenada |
| **Elemento del medio** | Mejor para listas ordenadas | No garantiza balance |
| **Aleatorio** | Evita peor caso | Overhead de generación aleatoria |
| **Mediana de 3** | Muy bueno en la práctica | Un poco más complejo |

**Mediana de 3:** Se eligen el primero, el del medio y el último, y se usa la mediana de estos tres como pivote

```python
# ============================================================
# QUICKSORT - ESQUEMA DE LOMUTO (Último elemento como pivote)
# ============================================================

def particion_lomuto(lista, bajo, alto):
    """
    Esquema de particionamiento de Lomuto.

    Elige el último elemento como pivote y reorganiza la lista de modo que:
    - Todos los elementos menores que el pivote queden a su izquierda
    - Todos los elementos mayores queden a su derecha

    Args:
        lista: Lista a particionar
        bajo: Índice inicial de la sublista
        alto: Índice final de la sublista

    Returns:
        Índice final del pivote
    """
    # Elegimos el último elemento como pivote
    pivote = lista[alto]

    # i marca la frontera: elementos antes de i son menores que el pivote
    i = bajo - 1

    # Recorremos desde 'bajo' hasta 'alto - 1'
    for j in range(bajo, alto):
        # Si el elemento actual es menor o igual al pivote
        if lista[j] <= pivote:
            i += 1  # Avanzamos la frontera
            # Intercambiamos lista[i] con lista[j]
            lista[i], lista[j] = lista[j], lista[i]

    # Colocamos el pivote en su posición correcta (justo después de la frontera)
    lista[i + 1], lista[alto] = lista[alto], lista[i + 1]

    # Retornamos la posición final del pivote
    return i + 1

def quicksort_lomuto(lista, bajo=None, alto=None):
    """
    QuickSort usando esquema de Lomuto.

    Complejidad:
        Tiempo: O(n log n) promedio, O(n²) peor caso
        Espacio: O(log n) - pila de recursión

    Args:
        lista: Lista a ordenar
        bajo: Índice inicial (por defecto 0)
        alto: Índice final (por defecto len(lista)-1)
    """
    # Si no se proporcionan, usamos los valores por defecto
    if bajo is None:
        bajo = 0
    if alto is None:
        alto = len(lista) - 1

    # Caso base: si la sublista tiene 0 o 1 elemento, ya está ordenada
    if bajo < alto:
        # Particionamos y obtenemos la posición del pivote
        pi = particion_lomuto(lista, bajo, alto)

        # Recursión: ordenamos las dos sublistas resultantes
        quicksort_lomuto(lista, bajo, pi - 1)      # Izquierda del pivote
        quicksort_lomuto(lista, pi + 1, alto)      # Derecha del pivote

    return lista

print("=== QUICKSORT CON LOMUTO ===\n")

# Ejemplo 1: Lista pequeña
lista1 = [8, 3, 1, 7, 0, 10, 2]
print(f"Lista original: {lista1}")
resultado1 = quicksort_lomuto(lista1.copy())
print(f"Lista ordenada: {resultado1}\n")

```

```python
# Ejemplo 2: Lista aleatoria
lista2 = [random.randint(1, 100) for _ in range(15)]
print(f"Lista aleatoria: {lista2}")
resultado2 = quicksort_lomuto(lista2.copy())
print(f"Lista ordenada: {resultado2}\n")

```

```python
# Ejemplo 3: Lista ya ordenada
lista3 = [1, 2, 3, 4, 5, 6, 7, 8]
print(f"Lista ordenada: {lista3}")
resultado3 = quicksort_lomuto(lista3.copy())
print(f"Resultado: {resultado3}\n")
```

```python
# Ejemplo 4: Lista con duplicados
lista4 = [5, 2, 8, 2, 5, 1, 8, 3]
print(f"Con duplicados: {lista4}")
resultado4 = quicksort_lomuto(lista4.copy())
print(f"Resultado: {resultado4}")
```

```python
# ============================================================
# QUICKSORT CON TRAZAS VISUALES
# ============================================================

def particion_con_trazas(lista, bajo, alto, profundidad=0):
    """Partición de Lomuto con trazas visuales."""
    indent = "  " * profundidad
    pivote = lista[alto]

    print(f"{indent} Pivote: {pivote} (posición {alto})")
    print(f"{indent} Sublista: {lista[bajo:alto+1]}")

    i = bajo - 1

    for j in range(bajo, alto):
        if lista[j] <= pivote:
            i += 1
            lista[i], lista[j] = lista[j], lista[i]
            print(f"{indent}  ↔ Intercambio: {lista[bajo:alto+1]}")

    lista[i + 1], lista[alto] = lista[alto], lista[i + 1]

    print(f"{indent} Pivote en posición {i+1}: {lista[bajo:alto+1]}")

    return i + 1

def quicksort_con_trazas(lista, bajo=None, alto=None, profundidad=0):
    """QuickSort con trazas visuales del proceso."""
    if bajo is None:
        bajo = 0
    if alto is None:
        alto = len(lista) - 1

    indent = "  " * profundidad

    # Caso base
    if bajo >= alto:
        if bajo == alto:
            print(f"{indent} Elemento único: {lista[bajo]}")
        return lista

    print(f"\n{indent}{'='*50}")
    print(f"{indent} QuickSort en posiciones [{bajo}, {alto}]")
    print(f"{indent} Lista actual: {lista}")

    # Particionamiento
    pi = particion_con_trazas(lista, bajo, alto, profundidad)

    # Recursión izquierda
    print(f"{indent} Recursión IZQUIERDA [{bajo}, {pi-1}]")
    quicksort_con_trazas(lista, bajo, pi - 1, profundidad + 1)

    # Recursión derecha
    print(f"{indent} Recursión DERECHA [{pi+1}, {alto}]")
    quicksort_con_trazas(lista, pi + 1, alto, profundidad + 1)

    return lista

print("=== TRAZANDO QUICKSORT ===\n")
lista_ejemplo = [8, 3, 1, 7, 0, 10, 2]
print(f"Lista inicial: {lista_ejemplo}\n")

quicksort_con_trazas(lista_ejemplo.copy())

print("\n" + "="*60)
print("Lista final ordenada:", lista_ejemplo)
```

```python
# QUICKSORT CON MEDIANA DE 3 (Mejor estrategia de pivote)

def mediana_de_tres(lista, bajo, alto):
    """
    Encuentra la mediana entre el primer, medio y último elemento
    Esta estrategia evita el peor caso en listas ya ordenadas
    """
    medio = (bajo + alto) // 2

    # Ordenamos los tres elementos
    if lista[bajo] > lista[medio]:
        lista[bajo], lista[medio] = lista[medio], lista[bajo]
    if lista[bajo] > lista[alto]:
        lista[bajo], lista[alto] = lista[alto], lista[bajo]
    if lista[medio] > lista[alto]:
        lista[medio], lista[alto] = lista[alto], lista[medio]

    # La mediana está en 'medio', la movemos a 'alto-1' para usarla como pivote
    lista[medio], lista[alto - 1] = lista[alto - 1], lista[medio]

    return lista[alto - 1]

def quicksort_mediana_de_3(lista, bajo=None, alto=None):
    """
    QuickSort usando mediana de 3 como pivote
    Más eficiente en la práctica que el esquema básico
    """
    if bajo is None:
        bajo = 0
    if alto is None:
        alto = len(lista) - 1

    # Caso base: listas pequeñas
    if bajo >= alto:
        return lista

    # Para listas pequeñas (<= 3 elementos), usamos inserción simple
    if alto - bajo < 3:
        # Ordenamiento por inserción para sublistas pequeñas
        for i in range(bajo + 1, alto + 1):
            clave = lista[i]
            j = i - 1
            while j >= bajo and lista[j] > clave:
                lista[j + 1] = lista[j]
                j -= 1
            lista[j + 1] = clave
        return lista

    # Elegimos pivote usando mediana de 3
    pivote = mediana_de_tres(lista, bajo, alto)

    # Particionamiento usando el pivote en posición alto-1
    i = bajo
    j = alto - 1

    while True:
        # Avanzamos i mientras sea menor que el pivote
        i += 1
        while lista[i] < pivote:
            i += 1

        # Retrocedemos j mientras sea mayor que el pivote
        j -= 1
        while lista[j] > pivote:
            j -= 1

        # Si los punteros se cruzaron, terminamos
        if i >= j:
            break

        # Intercambiamos
        lista[i], lista[j] = lista[j], lista[i]

    # Colocamos el pivote en su posición final
    lista[i], lista[alto - 1] = lista[alto - 1], lista[i]

    # Recursión
    quicksort_mediana_de_3(lista, bajo, i - 1)
    quicksort_mediana_de_3(lista, i + 1, alto)

    return lista

print("=== QUICKSORT CON MEDIANA DE 3 ===\n")

# lista ya ordenada
lista_ordenada = list(range(1, 11))
print(f"Lista ordenada: {lista_ordenada}")
resultado = quicksort_mediana_de_3(lista_ordenada.copy())
print(f"Resultado: {resultado}\n")

# Lista en orden inverso
lista_inversa = list(range(10, 0, -1))
print(f"Lista inversa: {lista_inversa}")
resultado = quicksort_mediana_de_3(lista_inversa.copy())
print(f"Resultado: {resultado}\n")

# Lista aleatoria
lista_random = [random.randint(1, 100) for _ in range(15)]
print(f"Lista aleatoria: {lista_random}")
resultado = quicksort_mediana_de_3(lista_random.copy())
print(f"Resultado: {resultado}")
```

##  Análisis de Complejidad de QuickSort

### Mejor Caso: O(n log n)

Ocurre cuando el particionamiento divide la lista en dos mitades **aproximadamente iguales** en cada paso

**Árbol de recursión:**
- Altura: log₂(n) niveles
- Trabajo por nivel: O(n) (recorrer todos los elementos)
- Total: O(n log n)


Ejemplo con n=8:

Nivel 0: [8 elementos] → partición

Nivel 1: [4 elementos] [4 elementos] → 2 particiones

Nivel 2: [2] [2] [2] [2] → 4 particiones

Nivel 3: [1] [1] [1] [1] [1] [1] [1] [1] → caso base


### Peor Caso: O(n²)

Ocurre cuando el particionamiento es **muy desbalanceado**, creando una sublista de n-1 elementos y otra de 0 elementos.

**Ejemplo:** Lista ya ordenada con pivote = último elemento


[1, 2, 3, 4, 5] → pivote=5

Partición: [1, 2, 3, 4] y [] → muy desbalanceado



**Árbol de recursión:**
- Altura: n niveles (lineal)
- Trabajo por nivel: O(n), O(n-1), O(n-2), ...
- Total: O(n²)

### Caso Promedio: O(n log n)

Con selección aleatoria del pivote o mediana de 3, la probabilidad de obtener particiones balanceadas es alta

- En promedio, las particiones dividen la lista en proporción 1:3 o mejor



```python
# COMPARACIÓN DE RENDIMIENTO: DIFERENTES ESTRATEGIAS DE PIVOTE

def medir_tiempo(funcion, lista, repeticiones=5):
    """Mide el tiempo promedio de ejecución."""
    tiempos = []
    for _ in range(repeticiones):
        lista_copia = lista.copy()
        inicio = time.time()
        funcion(lista_copia)
        fin = time.time()
        tiempos.append(fin - inicio)
    return sum(tiempos) / len(tiempos)

# Generamos diferentes tipos de listas
tamanos = [1000, 2000, 3000, 4000, 5000]

print(" COMPARACIÓN DE TIEMPOS\n")
print(f"{'n':>6} | {'Lomuto':>12} | {'Mediana de 3':>12} | {'Mejora':>10}")
print("-" * 50)

for n in tamanos:
    # Lista aleatoria
    lista_random = [random.randint(1, 10000) for _ in range(n)]

    t_lomuto = medir_tiempo(lambda l: quicksort_lomuto(l), lista_random)
    t_mediana = medir_tiempo(lambda l: quicksort_mediana_de_3(l), lista_random)

    mejora = (t_lomuto - t_mediana) / t_lomuto * 100

    print(f"{n:>6} | {t_lomuto:>12.6f} | {t_mediana:>12.6f} | {mejora:>+9.1f}%")

```

```python
# CASO PROBLEMÁTICO: LISTA YA ORDENADA

print("\n" + "="*60)
print("CASO PROBLEMÁTICO: Lista ya ordenada")
print("="*60 + "\n")

print(f"{'n':>6} | {'Lomuto':>12} | {'Mediana de 3':>12} | {'Mejora':>10}")
print("-" * 50)

for n in [1000, 2000, 3000, 4000, 5000]:
    lista_ordenada = list(range(n))

    # Aseguramos que Lomuto no falle para poder comparar
    # Aumentar el límite de recursión de Python temporalmente para Lomuto en este test

    import sys
    sys.setrecursionlimit(n + 100) # Ajustar según el tamaño máximo de n + un margen

    t_lomuto = medir_tiempo(lambda l: quicksort_lomuto(l), lista_ordenada)
    t_mediana = medir_tiempo(lambda l: quicksort_mediana_de_3(l), lista_ordenada)

    mejora = (t_lomuto - t_mediana) / t_lomuto * 100

    print(f"{n:>6} | {t_lomuto:>12.6f} | {t_mediana:>12.6f} | {mejora:>+9.1f}%")

print("• Lomuto sufre con listas ordenadas (peor caso O(n²))")
print("• Mediana de 3 maneja bien este caso (O(n log n))")
```

```python
# ============================================================
# DEMOSTRACIÓN DEL PEOR CASO
# ============================================================

print("=== DEMOSTRACIÓN DEL PEOR CASO ===\n")

# Lista ya ordenada - peor caso para Lomuto
lista_ordenada = list(range(1, 11))
print(f"Lista ordenada: {lista_ordenada}")
print("\nCon Lomuto (pivote = último elemento):")
print("Paso 1: Pivote=10, partición: [1,2,3,4,5,6,7,8,9] y []")
print("Paso 2: Pivote=9, partición: [1,2,3,4,5,6,7,8] y []")
print("...")
print("Paso 9: Pivote=2, partición: [1] y []")
print("\nTotal de comparaciones: 10+9+8+...+1 = 55 ≈ n²/2")
print("Complejidad: O(n²)\n")

# Contamos comparaciones reales
def quicksort_con_conteo(lista, bajo=None, alto=None):
    """QuickSort que cuenta comparaciones."""
    if bajo is None:
        bajo = 0
    if alto is None:
        alto = len(lista) - 1

    if bajo >= alto:
        return 0

    # Partición con conteo
    pivote = lista[alto]
    i = bajo - 1
    comparaciones = 0

    for j in range(bajo, alto):
        comparaciones += 1  # Contamos la comparación
        if lista[j] <= pivote:
            i += 1
            lista[i], lista[j] = lista[j], lista[i]

    lista[i + 1], lista[alto] = lista[alto], lista[i + 1]
    pi = i + 1

    # Recursión
    comparaciones += quicksort_con_conteo(lista, bajo, pi - 1)
    comparaciones += quicksort_con_conteo(lista, pi + 1, alto)

    return comparaciones

# Prueba
lista_test = list(range(1, 11))
comps_ordenada = quicksort_con_conteo(lista_test.copy())
print(f"Comparaciones con lista ordenada: {comps_ordenada}")
print(f"Fórmula: n(n-1)/2 = {10*9//2}")
print(f"Complejidad: O(n²)\n")

# Lista aleatoria
lista_random = [random.randint(1, 100) for _ in range(10)]
comps_random = quicksort_con_conteo(lista_random.copy())
print(f"Comparaciones con lista aleatoria: {comps_random}")
print(f"Complejidad esperada: O(n log n) ≈ {10 * np.log2(10):.1f}")
```

## QUICKSORT CON PIVOTE ALEATORIO

Una forma efectiva de evitar el peor caso de QuickSort (cuando el pivote siempre es el más pequeño o el más grande) es seleccionar el pivote de forma aleatoria.

```python
# ============================================================
# QUICKSORT - ESQUEMA DE LOMUTO CON PIVOTE ALEATORIO
# ============================================================

def particion_lomuto_aleatoria(lista, bajo, alto):
    """
    Esquema de Lomuto con selección de pivote aleatorio.
    Primero, selecciona un pivote aleatorio y lo intercambia con el último elemento.
    Luego, procede con el particionamiento estándar de Lomuto.
    """
    # Genera un índice aleatorio entre bajo y alto
    indice_aleatorio = random.randint(bajo, alto)

    # Intercambia el elemento aleatorio con el último elemento (el que Lomuto usará como pivote)
    lista[indice_aleatorio], lista[alto] = lista[alto], lista[indice_aleatorio]

    # Ahora llamamos a la partición de Lomuto estándar
    return particion_lomuto(lista, bajo, alto)

def quicksort_aleatorio(lista, bajo=None, alto=None):
    """
    QuickSort usando esquema de Lomuto con pivote aleatorio.

    Complejidad:
        Tiempo: O(n log n) promedio (evita el peor caso con alta probabilidad)
        Espacio: O(log n) - pila de recursión
    """
    if bajo is None:
        bajo = 0
    if alto is None:
        alto = len(lista) - 1

    if bajo < alto:
        # Usamos la partición con pivote aleatorio
        pi = particion_lomuto_aleatoria(lista, bajo, alto)

        quicksort_aleatorio(lista, bajo, pi - 1)
        quicksort_aleatorio(lista, pi + 1, alto)

    return lista

print("=== QUICKSORT CON PIVOTE ALEATORIO ===\n")

# Ejemplo 1: Lista pequeña
lista_ejemplo_aleatorio = [8, 3, 1, 7, 0, 10, 2]
print(f"Lista original: {lista_ejemplo_aleatorio}")
resultado_aleatorio = quicksort_aleatorio(lista_ejemplo_aleatorio.copy())
print(f"Lista ordenada: {resultado_aleatorio}\n")

# Ejemplo 2: Lista ya ordenada (anteriormente el peor caso para Lomuto)
lista_ordenada_peor_caso = list(range(1, 11))
print(f"Lista ordenada (peor caso para Lomuto): {lista_ordenada_peor_caso}")
resultado_ordenada_aleatorio = quicksort_aleatorio(lista_ordenada_peor_caso.copy())
print(f"Lista ordenada (con pivote aleatorio): {resultado_ordenada_aleatorio}\n")
```
