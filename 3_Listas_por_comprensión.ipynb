## ¿Qué es una lista por comprensión?

Una lista por comprensión es una forma corta y elegante de crear listas

En lugar de:

1. Crear una lista vacía
2. Recorrer elementos con un ciclo
3. Agregar resultados

Podemos hacerlo en una sola línea

```python
# La forma "tradicional de escribirlo"

numeros = [1,2,3,4,5]

cuadrados = []

for n in numeros:
    cuadrados.append(n**2)

print(cuadrados)
```

## La misma solución con Listas de comprensión

### Sintaxis general

[nueva_expresion for elemento in iterable]

La comprensión realiza:

- Recorrido
- Transformación
- Creación de la lista

Todo en una sola instrucción

```python
numeros = [1,2,3,4,5]

cuadrados = [n**2 for n in numeros]

print(cuadrados)
```

[n**2 for n in numeros]

n**2                → transformación

for n               → variable temporal

in numeros          → colección a recorrer

```python
frutas = ["manzana", "pera", "uva"]

mayusculas = [fruta.upper() for fruta in frutas]

print(mayusculas)
```

### Ejercicio 1

['ANA', 'LUIS', 'SOFIA']

a partir de:

nombres = ["Ana", "Luis", "Sofia"]

```python
nombres = ["Ana", "Luis", "Sofia"]

resultado = [nombre.upper() for nombre in nombres]

print(resultado)
```

# Agregando condiciones

También podemos filtrar elementos

Sintaxis:

[expresion for elemento in iterable if condicion]

```python
numeros = range(1,11)

pares = [n for n in numeros if n % 2 == 0]

print(pares)
```

```python
numeros = range(1,21)

multiplos_3 = [n for n in numeros if n % 3 == 0]

print(multiplos_3)
```

# Condicional IF-ELSE

También podemos transformar de manera diferente según una condición

Sintaxis:

[valor_si_true if condicion else valor_si_false for elemento in iterable]

```python
numeros = range(1,11)

resultado = ["Par" if n % 2 == 0 else "Impar" for n in numeros]

print(resultado)
```

```python
calificaciones = [95, 80, 55, 70, 40]

estatus = [
    "Aprobado" if calificacion >= 70 else "Reprobado"
    for calificacion in calificaciones
]

print(estatus)
```

Las listas por comprensión son muy utilizadas para:

- Limpieza de datos
- Transformación de variables
- Filtrado de registros
- Automatización de reportes

```python
ventas = [150000, 230000, 8000, 12000, 50000]

ventas_miles = [v / 1000 for v in ventas]

print(ventas_miles)
```

```python
ventas = [1500, 2300, 800, 1200, 500]

ventas_altas = [v for v in ventas if v > 1000]

print(ventas_altas)
```

# Comprensiones anidadas

Podemos usar múltiples ciclos dentro de una comprensión

```python
pares = [(x, y) for x in range(3) for y in range(3)]

print(pares)
```

## Transformación de valores en un diccionario anidado

Supongamos que queremos convertir ventas de pesos a miles de pesos:

```python
ventas = {
    "Norte": {"Enero": 15000, "Febrero": 18000},
    "Sur": {"Enero": 12000, "Febrero": 16000}
}

ventas_miles = {
    sucursal: {mes: valor / 1000 for mes, valor in meses.items()}
    for sucursal, meses in ventas.items()
}

print(ventas_miles)
```

## filtrar registros dentro de diccionarios anidados

Ahora queremos conservar únicamente ventas mayores a 15,000


```python
ventas_altas = {
sucursal: {mes: v for mes, v in meses.items() if v > 15000}
for sucursal, meses in ventas.items()
}
print(ventas_altas)
```
