# Algoritmos de Ordenamiento Recursivo

- **Recursión:** Función que se llama a sí misma
- **Caso base:** Condición que detiene la recursión
- **Paso recursivo:** Llamada con problema más pequeño
- **Pila de llamadas:** Memoria usada por cada llamada recursiva

## ¿Qué es la Recursión?

La **recursión** es una técnica donde una función **se llama a sí misma** para resolver un problema dividiéndolo en subproblemas más pequeños

1. **Caso base:** Condición que detiene la recursión (evita bucle infinito)
2. **Paso recursivo:** La función se llama a sí misma con un problema más pequeño
3. **Reducción:** Cada llamada debe acercarse al caso base


Imagina una muñeca rusa:
- Abres la muñeca grande, hay una más pequeña dentro
- Abres la muñeca mediana, hay una más pequeña dentro
- ...
- Abres la muñeca más pequeña → **no hay nada dentro** (CASO BASE)
- Luego cierras todas en orden inverso

### Ejemplo clásico: Factorial

```python
def factorial(n):
    # CASO BASE
    if n <= 1:
        return 1
    # PASO RECURSIVO
    return n * factorial(n - 1)

# factorial(4) = 4 * factorial(3)
#              = 4 * 3 * factorial(2)
#              = 4 * 3 * 2 * factorial(1)
#              = 4 * 3 * 2 * 1
#              = 24

¿Por qué usar recursión en ordenamiento?

Aunque los algoritmos de ordenamiento simples (Inserción, Selección) son naturalmente iterativos, podemos reinterpretarlos recursivamente para:

Entender mejor el patrón "divide y vencerás"

Prepararnos para algoritmos más complejos (MergeSort, QuickSort)

Ver cómo la recursión afecta el rendimiento (pila de llamadas)


## Insertion Sort

Imagina que estás jugando cartas y quieres ordenar tu mano:

1. **Primera carta:** Ya está "ordenada" (solo hay una)
2. **Segunda carta:** La comparas con la primera y la colocas en su posición
3. **Tercera carta:** La comparas con las dos primeras y la insertas en su lugar
4. **Cuarta carta:** La comparas con las tres primeras y la insertas...
5. **Continúas** hasta haber insertado todas las cartas

Lista: `[5, 2, 4, 6, 1, 3]`

Estado inicial: [5] | [2, 4, 6, 1, 3]

↑

Ya ordenada

Paso 1: Insertar 2 en [5]

→ [2, 5] | [4, 6, 1, 3]

Paso 2: Insertar 4 en [2, 5]

→ [2, 4, 5] | [6, 1, 3]

Paso 3: Insertar 6 en [2, 4, 5]

→ [2, 4, 5, 6] | [1, 3]

Paso 4: Insertar 1 en [2, 4, 5, 6]

→ [1, 2, 4, 5, 6] | [3]

Paso 5: Insertar 3 en [1, 2, 4, 5, 6]

→ [1, 2, 3, 4, 5, 6] Ordenado


### Patrón recursivo

Para ordenar `n` elementos:
1. **Ordena los primeros `n-1` elementos** (llamada recursiva)
2. **Inserta el n-ésimo elemento** en su posición correcta (paso iterativo)

**Caso base:** Si `n ≤ 1`, la lista ya está ordenada


````python
# ============================================================
# INSERTION SORT - VERSIÓN RECURSIVA
# ============================================================

def insertion_sort_recursivo(lista, n=None):
    """
    Ordena una lista usando Insertion Sort de forma recursiva

    ESTRATEGIA RECURSIVA:
    1. Caso base: si n <= 1, la lista ya está ordenada
    2. Paso recursivo: ordenar los primeros n-1 elementos
    3. Insertar el n-ésimo elemento en su posición correcta

    Args:
        lista: Lista de números a ordenar
        n: Tamaño de la sublista a ordenar (por defecto, toda la lista)

    Returns:
        La misma lista ordenada (modifica la lista original)
    """
    # Si n no se proporciona, usamos el tamaño completo
    if n is None:
        n = len(lista)

    # CASO BASE: Lista de 0 o 1 elemento ya está ordenada

    if n <= 1:
        return lista

    # PASO RECURSIVO: Primero ordenamos los primeros n-1 elementos
    insertion_sort_recursivo(lista, n - 1)

    # PASO DE INSERCIÓN: Insertar lista[n-1] en su posición correcta

    # Guardamos el elemento que vamos a insertar
    clave = lista[n - 1]

    # Empezamos desde el elemento anterior al último
    j = n - 2

    # Desplazamos elementos mayores que 'clave' hacia la derecha
    while j >= 0 and lista[j] > clave:
        lista[j + 1] = lista[j]  # Movemos elemento una posición a la derecha
        j -= 1

    # Insertamos la clave en su posición correcta
    lista[j + 1] = clave

    return lista

# PRUEBA
print("=== INSERTION SORT RECURSIVO ===\n")

# Ejemplo 1: Lista pequeña
lista1 = [5, 2, 4, 6, 1, 3]
print(f"Lista original: {lista1}")
resultado1 = insertion_sort_recursivo(lista1.copy())
print(f"Lista ordenada: {resultado1}\n")

# Ejemplo 2: Lista aleatoria
lista2 = [random.randint(1, 100) for _ in range(10)]
print(f"Lista aleatoria: {lista2}")
resultado2 = insertion_sort_recursivo(lista2.copy())
print(f"Lista ordenada: {resultado2}\n")

# Ejemplo 3: Lista ya ordenada (mejor caso)
lista3 = [1, 2, 3, 4, 5, 6]
print(f"Lista ya ordenada: {lista3}")
resultado3 = insertion_sort_recursivo(lista3.copy())
print(f"Resultado: {resultado3}\n")

# Ejemplo 4: Lista en orden inverso (peor caso)
lista4 = [6, 5, 4, 3, 2, 1]
print(f"Lista en orden inverso: {lista4}")
resultado4 = insertion_sort_recursivo(lista4.copy())
print(f"Resultado: {resultado4}")
````

3. **Pila de llamadas:** En el punto más profundo, hay `n` llamadas apiladas en memoria


## Selection Sort

Imagina que tienes una fila de libros de diferentes alturas y quieres ordenarlos de menor a mayor:

1. **Busca el libro más bajo** de toda la fila
2. **Intercámbialo** con el libro en la primera posición
3. **Busca el libro más bajo** entre los restantes (desde la segunda posición)
4. **Intercámbialo** con el libro en la segunda posición
5. **Repite** hasta que todos estén ordenados

Lista: `[64, 25, 12, 22, 11]`


Estado inicial: [64, 25, 12, 22, 11]

Paso 1: Mínimo = 11 (posición 4)

Intercambiar con posición 0

→ [11, 25, 12, 22, 64]

Paso 2: Mínimo = 12 (posición 2)

Intercambiar con posición 1

→ [11, 12, 25, 22, 64]

Paso 3: Mínimo = 22 (posición 3)

Intercambiar con posición 2

→ [11, 12, 22, 25, 64]

Paso 4: Mínimo = 25 (posición 3)

Ya está en su lugar

→ [11, 12, 22, 25, 64]



Para ordenar desde la posición `inicio`:
1. **Encuentra el mínimo** desde `inicio` hasta el final (paso iterativo)
2. **Intercámbialo** con `lista[inicio]`
3. **Ordena el resto** recursivamente desde `inicio + 1`

**Caso base:** Si `inicio >= n - 1`, la lista ya está ordenada

````python
# SELECTION SORT - VERSIÓN RECURSIVA

def selection_sort_recursivo(lista, inicio=0):
    """
    Ordena una lista usando Selection Sort de forma recursiva.

    ESTRATEGIA RECURSIVA:
    1. Caso base: si inicio >= len(lista) - 1, la lista ya está ordenada
    2. Paso iterativo: encontrar el mínimo desde 'inicio' hasta el final
    3. Intercambiar el mínimo con lista[inicio]
    4. Paso recursivo: ordenar el resto desde inicio + 1

    Args:
        lista: Lista de números a ordenar
        inicio: Índice desde donde comienza la sublista no ordenada

    Returns:
        La misma lista ordenada (modifica la lista original)
    """
    n = len(lista)

    # CASO BASE: Si llegamos al último elemento, la lista está ordenada

    if inicio >= n - 1:
        return lista

    # PASO ITERATIVO: Encontrar el índice del mínimo
    indice_minimo = inicio

    for j in range(inicio + 1, n):
        if lista[j] < lista[indice_minimo]:
            indice_minimo = j

    # INTERCAMBIO: Poner el mínimo en su posición correcta
    if indice_minimo != inicio:
        lista[inicio], lista[indice_minimo] = lista[indice_minimo], lista[inicio]

    # PASO RECURSIVO: Ordenar el resto de la lista
    return selection_sort_recursivo(lista, inicio + 1)

# PRUEBA

# Ejemplo 1: Lista pequeña
lista1 = [64, 25, 12, 22, 11]
print(f"Lista original: {lista1}")
resultado1 = selection_sort_recursivo(lista1.copy())
print(f"Lista ordenada: {resultado1}\n")

# Ejemplo 2: Lista aleatoria
lista2 = [random.randint(1, 100) for _ in range(10)]
print(f"Lista aleatoria: {lista2}")
resultado2 = selection_sort_recursivo(lista2.copy())
print(f"Lista ordenada: {resultado2}\n")

# Ejemplo 3: Lista ya ordenada
lista3 = [1, 2, 3, 4, 5, 6]
print(f"Lista ya ordenada: {lista3}")
resultado3 = selection_sort_recursivo(lista3.copy())
print(f"Resultado: {resultado3}\n")

# Ejemplo 4: Lista en orden inverso
lista4 = [6, 5, 4, 3, 2, 1]
print(f"Lista en orden inverso: {lista4}")
resultado4 = selection_sort_recursivo(lista4.copy())
print(f"Resultado: {resultado4}")
````

Diferencia con Insertion Sort

- **Insertion Sort recursivo:** La recursión ocurre ANTES de la inserción (primero ordena n-1, luego inserta el n-ésimo)
- **Selection Sort recursivo:** La recursión ocurre DESPUÉS del intercambio (primero encuentra e intercambia el mínimo, luego ordena el resto)


## Comparación: Recursivo vs Iterativo

###  Tabla Comparativa

| Aspecto | Versión Iterativa | Versión Recursiva |
|---------|-------------------|-------------------|
| **Estructura** | Bucles `for`/`while` | Llamadas a sí misma |
| **Caso base** | Condición del bucle | Condición `if` al inicio |
| **Memoria** | O(1) - constante | O(n) - pila de llamadas |
| **Velocidad** | Más rápida | Ligeramente más lenta |
| **Legibilidad** | Más directa | Más elegante conceptualmente |
| **Riesgo** | Ninguno | Stack Overflow si n es muy grande |

###  ¿Cuándo usar cada versión?

####  Usa **Iterativo** cuando:
- El rendimiento es crítico
- La lista es muy grande (evita stack overflow)
- Quieres mínimo uso de memoria
- Es la opción por defecto en producción

####  Usa **Recursivo** cuando:
- Estás aprendiendo conceptos de recursión
- El problema tiene estructura naturalmente recursiva
- Quieres código más elegante y matemático
- Es preparación para algoritmos como MergeSort/QuickSort

###  El Problema del Stack Overflow

Python tiene un límite de recursión (por defecto ~1000 llamadas):

```python
import sys
print(sys.getrecursionlimit())  # 1000

# Si intentas ordenar una lista de 2000 elementos con recursión:
# RecursionError: maximum recursion depth exceeded

Analogía Final

Imagina que quieres subir una escalera de 10 escalones:

Iterativo: Subes escalón por escalón con un bucle

````python
def subo_un_escalón():
  print("Subiendo un escalón...")

for escalon in range(10):
  subo_un_escalón()
````

Recursivo: Para llegar al escalón 10, primero llegas al 9, que requiere llegar al 8...

````python
  def subir(escalón):
      if escalón == 0:
          return  # Caso base
      subir(escalón - 1)  # Llamada recursiva
      subo_un_escalón()
````

Ambos llegan arriba, pero el recursivo necesita recordar todos los escalones anteriores en su "memoria" (pila de llamadas).

````python
import sys

# ============================================================
# COMPARACIÓN DE RENDIMIENTO: ITERATIVO vs RECURSIVO
# ============================================================

# Primero, implementamos las versiones iterativas para comparar

def insertion_sort_iterativo(lista):
    """Versión iterativa de Insertion Sort."""
    n = len(lista)
    for i in range(1, n):
        clave = lista[i]
        j = i - 1
        while j >= 0 and lista[j] > clave:
            lista[j + 1] = lista[j]
            j -= 1
        lista[j + 1] = clave
    return lista

def selection_sort_iterativo(lista):
    """Versión iterativa de Selection Sort."""
    n = len(lista)
    for i in range(n - 1):
        indice_minimo = i
        for j in range(i + 1, n):
            if lista[j] < lista[indice_minimo]:
                indice_minimo = j
        if indice_minimo != i:
            lista[i], lista[indice_minimo] = lista[indice_minimo], lista[i]
    return lista

def insertion_sort_recursivo(lista, n=None):
    """Versión recursiva de Insertion Sort."""
    if n is None:
        n = len(lista)
    if n <= 1:
        return lista
    insertion_sort_recursivo(lista, n - 1)
    clave = lista[n - 1]
    j = n - 2
    while j >= 0 and lista[j] > clave:
        lista[j + 1] = lista[j]
        j -= 1
    lista[j + 1] = clave
    return lista

def selection_sort_recursivo(lista, inicio=0):
    """Versión recursiva de Selection Sort."""
    n = len(lista)
    if inicio >= n - 1:
        return lista
    indice_minimo = inicio
    for j in range(inicio + 1, n):
        if lista[j] < lista[indice_minimo]:
            indice_minimo = j
    if indice_minimo != inicio:
        lista[inicio], lista[indice_minimo] = lista[indice_minimo], lista[inicio]
    return selection_sort_recursivo(lista, inicio + 1)

# MEDICIÓN DE TIEMPOS

def medir_tiempo(funcion, lista, repeticiones=5):
    """Mide el tiempo promedio de ejecutar una función."""
    tiempos = []
    for _ in range(repeticiones):
        lista_copia = lista.copy()
        inicio = time.time()
        funcion(lista_copia)
        fin = time.time()
        tiempos.append(fin - inicio)
    return sum(tiempos) / len(tiempos)

# Generamos listas de diferentes tamaños
tamanos = [100, 500, 1000, 2000, 3000, 4000]

tiempos_ins_iter = []
tiempos_ins_rec = []
tiempos_sel_iter = []
tiempos_sel_rec = []

print(" COMPARACIÓN DE TIEMPOS DE EJECUCIÓN\n")
print(f"{'n':>6} | {'Ins.Iter':>10} | {'Ins.Rec':>10} | {'Sel.Iter':>10} | {'Sel.Rec':>10} | {'Diferencia':>10}")
print("-" * 75)

# Guardamos el límite de recursión original
original_recursion_limit = sys.getrecursionlimit()

for n in tamanos:
    # Aumentamos el límite de recursión para n grandes
    # Aseguramos que sea al menos n + un margen, sin reducirlo si ya es mayor
    sys.setrecursionlimit(max(original_recursion_limit, n + 50))

    lista = [random.randint(1, 10000) for _ in range(n)]

    t_ins_iter = medir_tiempo(insertion_sort_iterativo, lista)
    t_ins_rec = medir_tiempo(insertion_sort_recursivo, lista)
    t_sel_iter = medir_tiempo(selection_sort_iterativo, lista)
    t_sel_rec = medir_tiempo(selection_sort_recursivo, lista)

    tiempos_ins_iter.append(t_ins_iter)
    tiempos_ins_rec.append(t_ins_rec)
    tiempos_sel_iter.append(t_sel_iter)
    tiempos_sel_rec.append(t_sel_rec)

    # Calculamos la diferencia promedio
    diff_promedio = ((t_ins_rec - t_ins_iter) + (t_sel_rec - t_sel_iter)) / 2

    print(f"{n:>6} | {t_ins_iter:>10.6f} | {t_ins_rec:>10.6f} | {t_sel_iter:>10.6f} | {t_sel_rec:>10.6f} | {diff_promedio:>+10.6f}")

# Restauramos el límite de recursión a su valor original al final
sys.setrecursionlimit(original_recursion_limit)
````

````python

# DEMOSTRACIÓN DEL MEJOR CASO DE INSERTION SORT

print("\n" + "="*60)
print("DEMOSTRACIÓN: Mejor caso de Insertion Sort O(n)")
print("="*60 + "\n")

lista_ordenada = list(range(1, 11))
print(f"Lista ya ordenada: {lista_ordenada}")

# Contamos comparaciones
def insertion_sort_con_conteo(lista, n=None):
    if n is None:
        n = len(lista)
    if n <= 1:
        return lista, 0

    lista, comparaciones_previas = insertion_sort_con_conteo(lista, n - 1)

    clave = lista[n - 1]
    j = n - 2
    comparaciones = 0

    while j >= 0 and lista[j] > clave:
        comparaciones += 1
        lista[j + 1] = lista[j]
        j -= 1

    comparaciones += 1  # La comparación que falla y sale del while
    lista[j + 1] = clave

    return lista, comparaciones_previas + comparaciones

_, comps_ordenada = insertion_sort_con_conteo(lista_ordenada.copy())
print(f"Comparaciones con lista ordenada: {comps_ordenada}")
print(f"→ Complejidad: O(n) = {len(lista_ordenada)} comparaciones\n")

lista_inversa = list(range(10, 0, -1))
print(f"Lista en orden inverso: {lista_inversa}")
_, comps_inversa = insertion_sort_con_conteo(lista_inversa.copy())
print(f"Comparaciones con lista inversa: {comps_inversa}")
print(f"→ Complejidad: O(n²) ≈ {len(lista_inversa)**2} comparaciones")
````
